```python
# ============================================================
# CELL 1.9: MRI GEOMETRY AUDIT
# Purpose:
#   Time1 এবং Time2 MRI + segmentation একই spatial geometry-তে
#   আছে কিনা check করা
#
# আমরা এখানে MRI পরিবর্তন করছি না — শুধু verify করছি.
# ============================================================

import nibabel as nib
import numpy as np
from tqdm.auto import tqdm

geometry_problems = []

for pid in tqdm(
    patient_ids,
    desc="Checking MRI geometry"
):

    folder = os.path.join(DATA_DIR, pid)

    paths = [
        os.path.join(folder, f"{pid}_time1_flair.nii.gz"),
        os.path.join(folder, f"{pid}_time1_t1ce.nii.gz"),
        os.path.join(folder, f"{pid}_time1_seg.nii.gz"),

        os.path.join(folder, f"{pid}_time2_flair.nii.gz"),
        os.path.join(folder, f"{pid}_time2_t1ce.nii.gz"),
        os.path.join(folder, f"{pid}_time2_seg.nii.gz")
    ]

    nii_objects = [
        nib.load(path)
        for path in paths
    ]

    shapes = [
        nii.shape
        for nii in nii_objects
    ]

    affines = [
        nii.affine
        for nii in nii_objects
    ]

    same_shape = all(
        shape == shapes[0]
        for shape in shapes
    )

    same_affine = all(
        np.allclose(
            affine,
            affines[0],
            atol=1e-3
        )
        for affine in affines
    )

    if not same_shape or not same_affine:

        geometry_problems.append({
            "PatientID": pid,
            "Shapes": shapes,
            "SameShape": same_shape,
            "SameAffine": same_affine
        })


print("\nGeometry problems:", len(geometry_problems))

if len(geometry_problems) == 0:
    print("✅ All MRI/SEG files have matching geometry.")
else:
    print("❌ Geometry mismatch detected.")

    for problem in geometry_problems[:5]:
        print(problem)
```

```python
# ============================================================
# CELL 3.1: LOAD ET SEGMENTATION MASKS
#
# Purpose:
#   Time1 এবং Time2 expert segmentation load করা
#   এবং Label 3 = Enhancing Tumor (ET) mask তৈরি করা
#
#   Visualization-এর জন্য যে slice-এ ET সবচেয়ে বেশি আছে,
#   সেটি automatically select করা.
#
# IMPORTANT:
#   Mask এখানে শুধু visualization/ground-truth analysis-এর জন্য।
#   এটি prediction model-এর input হিসেবে ব্যবহার করা হবে না.
# ============================================================

import os
import nibabel as nib
import numpy as np


# -----------------------------
# Segmentation paths
# -----------------------------

seg1_path = os.path.join(
    patient_folder,
    f"{patient_id}_time1_seg.nii.gz"
)

seg2_path = os.path.join(
    patient_folder,
    f"{patient_id}_time2_seg.nii.gz"
)


# -----------------------------
# Load segmentation volumes
# -----------------------------

seg1 = nib.load(seg1_path).get_fdata()
seg2 = nib.load(seg2_path).get_fdata()


# -----------------------------
# Enhancing Tumor = Label 3
# -----------------------------

ET_LABEL = 3

et1 = (seg1 == ET_LABEL)
et2 = (seg2 == ET_LABEL)


# -----------------------------
# Find the most informative slice
# using ET from BOTH timepoints
# -----------------------------

et_union_per_slice = (
    et1.sum(axis=(0, 1)) +
    et2.sum(axis=(0, 1))
)

best_z = int(
    np.argmax(et_union_per_slice)
)


print("Patient:", patient_id)
print("Best ET slice:", best_z)

print(
    "Time1 ET voxels:",
    int(et1[:, :, best_z].sum())
)

print(
    "Time2 ET voxels:",
    int(et2[:, :, best_z].sum())
)
```

```python
# ============================================================
# CELL 3.3: ACTUAL TUMOR EVOLUTION MAP
#
# Purpose:
#   Time1 এবং Time2 expert ET masks compare করে বের করা:
#
#   Growth     = Time2-তে tumor আছে, Time1-এ ছিল না
#   Regression = Time1-এ tumor ছিল, Time2-তে নেই
#   Stable     = দুই timepoint-এই tumor আছে
#
# এই actual tumor evolution পরে Grad-CAM attention change-এর
# সাথে quantitatively compare করা হবে.
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch


# ------------------------------------------------------------
# 1. Calculate voxel-wise tumor evolution
# ------------------------------------------------------------

growth = et2 & (~et1)

regression = et1 & (~et2)

stable = et1 & et2


# ------------------------------------------------------------
# 2. Whole 3D volume statistics
# ------------------------------------------------------------

growth_voxels = int(growth.sum())
regression_voxels = int(regression.sum())
stable_voxels = int(stable.sum())

time1_total = int(et1.sum())
time2_total = int(et2.sum())

net_change = time2_total - time1_total


print("========== 3D ET EVOLUTION ==========")

print("Time1 total ET voxels :", time1_total)
print("Time2 total ET voxels :", time2_total)

print("\nStable ET voxels      :", stable_voxels)
print("Growth voxels         :", growth_voxels)
print("Regression voxels     :", regression_voxels)

print("\nNet ET voxel change   :", net_change)


# ------------------------------------------------------------
# 3. Select best slice from previous cell
# ------------------------------------------------------------

growth_slice = growth[:, :, best_z]
regression_slice = regression[:, :, best_z]
stable_slice = stable[:, :, best_z]


# ------------------------------------------------------------
# 4. Create combined evolution map
#
# 0 = background
# 1 = stable
# 2 = growth
# 3 = regression
# ------------------------------------------------------------

evolution_map = np.zeros(
    et1[:, :, best_z].shape,
    dtype=np.uint8
)

evolution_map[stable_slice] = 1
evolution_map[growth_slice] = 2
evolution_map[regression_slice] = 3


# ------------------------------------------------------------
# 5. Plot individual regions
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2,
    2,
    figsize=(11, 10)
)


# Stable
axes[0, 0].imshow(
    stable_slice,
    cmap="gray"
)

axes[0, 0].set_title(
    "Stable ET Region"
)

axes[0, 0].axis("off")


# Growth
axes[0, 1].imshow(
    growth_slice,
    cmap="gray"
)

axes[0, 1].set_title(
    "ET Growth Region"
)

axes[0, 1].axis("off")


# Regression
axes[1, 0].imshow(
    regression_slice,
    cmap="gray"
)

axes[1, 0].set_title(
    "ET Regression Region"
)

axes[1, 0].axis("off")


# ------------------------------------------------------------
# 6. Combined evolution overlay on Time2 T1CE
# ------------------------------------------------------------

axes[1, 1].imshow(
    t2_t1ce[:, :, best_z],
    cmap="gray"
)

masked_evolution = np.ma.masked_where(
    evolution_map == 0,
    evolution_map
)

evolution_cmap = ListedColormap([
    "blue",     # Stable
    "red",      # Growth
    "lime"      # Regression
])

axes[1, 1].imshow(
    masked_evolution,
    cmap=evolution_cmap,
    vmin=1,
    vmax=3,
    alpha=0.55
)

axes[1, 1].set_title(
    "Combined ET Evolution on Time2 T1CE"
)

axes[1, 1].axis("off")


# ------------------------------------------------------------
# 7. Legend
# ------------------------------------------------------------

legend_elements = [
    Patch(facecolor="blue", label="Stable"),
    Patch(facecolor="red", label="Growth"),
    Patch(facecolor="lime", label="Regression")
]

axes[1, 1].legend(
    handles=legend_elements,
    loc="lower right"
)


plt.suptitle(
    f"Patient {patient_id} — Actual ET Evolution\nSlice {best_z}",
    fontsize=15
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CELL 4.1: MRI PREPROCESSING TEST
#
# Purpose:
#   1. Black/background region identify করা
#   2. MRI intensity outlier clip করা
#   3. Non-zero brain voxels দিয়ে Z-score normalization করা
#   4. Brain region crop করা
#   5. 224×224 resize করা
#
# IMPORTANT:
#   Tumor segmentation mask preprocessing decision নিতে
#   ব্যবহার করা হচ্ছে না -> no tumor-mask leakage.
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import zoom


IMAGE_SIZE = 224


# ------------------------------------------------------------
# 1. MRI INTENSITY NORMALIZATION
# ------------------------------------------------------------

def normalize_mri(volume):

    volume = volume.astype(np.float32)

    # Dataset skull-stripped, therefore non-zero voxels ≈ brain
    brain_mask = volume != 0

    if brain_mask.sum() == 0:
        return volume

    values = volume[brain_mask]

    # Clip extreme intensity values
    lower = np.percentile(values, 0.5)
    upper = np.percentile(values, 99.5)

    clipped = np.clip(
        volume,
        lower,
        upper
    )

    mean = clipped[brain_mask].mean()
    std = clipped[brain_mask].std()

    normalized = np.zeros_like(
        clipped,
        dtype=np.float32
    )

    normalized[brain_mask] = (
        clipped[brain_mask] - mean
    ) / (std + 1e-6)

    return normalized


# ------------------------------------------------------------
# 2. FIND BRAIN BOUNDING BOX
#
# Time1 + Time2 + FLAIR + T1CE union ব্যবহার করছি,
# যাতে paired scans একই crop পায়.
# ------------------------------------------------------------

def get_brain_bbox(volumes, padding=5):

    union = np.zeros(
        volumes[0].shape,
        dtype=bool
    )

    for volume in volumes:
        union |= (volume != 0)

    coords = np.argwhere(union)

    x_min, y_min, z_min = coords.min(axis=0)
    x_max, y_max, z_max = coords.max(axis=0)

    x_min = max(0, x_min - padding)
    y_min = max(0, y_min - padding)
    z_min = max(0, z_min - padding)

    x_max = min(
        union.shape[0] - 1,
        x_max + padding
    )

    y_max = min(
        union.shape[1] - 1,
        y_max + padding
    )

    z_max = min(
        union.shape[2] - 1,
        z_max + padding
    )

    return (
        x_min, x_max,
        y_min, y_max,
        z_min, z_max
    )


# ------------------------------------------------------------
# 3. RESIZE A 2D SLICE TO 224×224
# ------------------------------------------------------------

def resize_slice(slice_2d, order=1):

    h, w = slice_2d.shape

    return zoom(
        slice_2d,
        (
            IMAGE_SIZE / h,
            IMAGE_SIZE / w
        ),
        order=order
    )


# ------------------------------------------------------------
# 4. NORMALIZE OUR FOUR MRI VOLUMES
# ------------------------------------------------------------

n_t1_flair = normalize_mri(t1_flair)
n_t2_flair = normalize_mri(t2_flair)

n_t1_t1ce = normalize_mri(t1_t1ce)
n_t2_t1ce = normalize_mri(t2_t1ce)


# ------------------------------------------------------------
# 5. ONE COMMON CROP FOR BOTH TIMEPOINTS
# ------------------------------------------------------------

bbox = get_brain_bbox([
    t1_flair,
    t2_flair,
    t1_t1ce,
    t2_t1ce
])

x1, x2, y1, y2, z1, z2 = bbox

print("Original shape:", t1_flair.shape)

print(
    "Brain bounding box:",
    bbox
)

print(
    "Usable axial range:",
    z1,
    "to",
    z2
)


# ------------------------------------------------------------
# 6. TEST USING PREVIOUS BEST SLICE
# ------------------------------------------------------------

raw_slice = t1_t1ce[:, :, best_z]

cropped_slice = n_t1_t1ce[
    x1:x2+1,
    y1:y2+1,
    best_z
]

processed_slice = resize_slice(
    cropped_slice,
    order=1
)


# ------------------------------------------------------------
# 7. VISUALIZE BEFORE vs AFTER
# ------------------------------------------------------------

plt.figure(figsize=(12, 4))


plt.subplot(1, 3, 1)

plt.imshow(
    raw_slice,
    cmap="gray"
)

plt.title(
    "Raw T1CE\n240×240"
)

plt.axis("off")


plt.subplot(1, 3, 2)

plt.imshow(
    cropped_slice,
    cmap="gray"
)

plt.title(
    "Normalized + Brain Crop"
)

plt.axis("off")


plt.subplot(1, 3, 3)

plt.imshow(
    processed_slice,
    cmap="gray"
)

plt.title(
    "Final CNN Input\n224×224"
)

plt.axis("off")


plt.suptitle(
    f"Patient {patient_id} — MRI Preprocessing Test",
    fontsize=14
)

plt.tight_layout()
plt.show()


print(
    "\nFinal processed slice shape:",
    processed_slice.shape
)

print(
    "Brain normalized mean:",
    round(
        n_t1_t1ce[
            n_t1_t1ce != 0
        ].mean(),
        4
    )
)

print(
    "Brain normalized std:",
    round(
        n_t1_t1ce[
            n_t1_t1ce != 0
        ].std(),
        4
    )
)
```

```python
# ============================================================
# CELL 4.2: ASPECT-RATIO PRESERVING MRI PREPROCESSING
#
# Purpose:
#   Brain crop-কে directly stretch না করে প্রথমে square padding
#   দেওয়া, তারপর 224×224 resize করা।
#
# এতে anatomical proportion preserve হবে এবং পরে
# Grad-CAM / segmentation spatial comparison reliable থাকবে.
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import zoom


# ------------------------------------------------------------
# 1. PAD A 2D IMAGE TO SQUARE
# ------------------------------------------------------------

def pad_to_square(image):

    h, w = image.shape

    size = max(h, w)

    padded = np.zeros(
        (size, size),
        dtype=image.dtype
    )

    top = (size - h) // 2
    left = (size - w) // 2

    padded[
        top:top+h,
        left:left+w
    ] = image

    return padded


# ------------------------------------------------------------
# 2. RESIZE SQUARE IMAGE TO 224 × 224
# ------------------------------------------------------------

def resize_to_224(image, order=1):

    h, w = image.shape

    resized = zoom(
        image,
        (
            IMAGE_SIZE / h,
            IMAGE_SIZE / w
        ),
        order=order
    )

    return resized


# ------------------------------------------------------------
# 3. TAKE THE SAME NORMALIZED + CROPPED TEST SLICE
# ------------------------------------------------------------

cropped_slice = n_t1_t1ce[
    x1:x2+1,
    y1:y2+1,
    best_z
]


# ------------------------------------------------------------
# 4. PAD TO SQUARE
# ------------------------------------------------------------

padded_slice = pad_to_square(
    cropped_slice
)


# ------------------------------------------------------------
# 5. FINAL 224 × 224 CNN INPUT
# ------------------------------------------------------------

final_slice = resize_to_224(
    padded_slice,
    order=1
)


# ------------------------------------------------------------
# 6. VISUALIZE ALL STEPS
# ------------------------------------------------------------

plt.figure(figsize=(15, 4))


# Raw
plt.subplot(1, 4, 1)

plt.imshow(
    t1_t1ce[:, :, best_z],
    cmap="gray"
)

plt.title("Raw MRI")
plt.axis("off")


# Cropped
plt.subplot(1, 4, 2)

plt.imshow(
    cropped_slice,
    cmap="gray"
)

plt.title(
    f"Brain Crop\n{cropped_slice.shape}"
)

plt.axis("off")


# Square padded
plt.subplot(1, 4, 3)

plt.imshow(
    padded_slice,
    cmap="gray"
)

plt.title(
    f"Square Padding\n{padded_slice.shape}"
)

plt.axis("off")


# Final
plt.subplot(1, 4, 4)

plt.imshow(
    final_slice,
    cmap="gray"
)

plt.title(
    "Final CNN Input\n224×224"
)

plt.axis("off")


plt.suptitle(
    f"Patient {patient_id} — Final MRI Spatial Preprocessing",
    fontsize=14
)

plt.tight_layout()
plt.show()


print("Original MRI shape :", t1_t1ce[:, :, best_z].shape)

print(
    "Cropped shape      :",
    cropped_slice.shape
)

print(
    "Padded shape       :",
    padded_slice.shape
)

print(
    "Final shape        :",
    final_slice.shape
)
```

```python
# ============================================================
# CELL 4.3: SEGMENTATION MASK PREPROCESSING
#
# Purpose:
#   MRI-এর জন্য যে SAME crop + square padding + resize ব্যবহার
#   করেছি, segmentation mask-এর উপরও exactly সেটাই apply করা।
#
# IMPORTANT:
#   Segmentation mask resize করার সময় order=0 ব্যবহার করছি
#   (nearest-neighbour interpolation), যাতে label values
#   0,1,2,3,4 পরিবর্তন না হয়।
#
# Final goal:
#   Processed MRI এবং processed expert mask pixel-to-pixel
#   aligned থাকবে Grad-CAM / IoU / Dice analysis-এর জন্য।
# ============================================================

import numpy as np
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. FUNCTION: APPLY SAME SPATIAL TRANSFORM TO SEGMENTATION
# ------------------------------------------------------------

def preprocess_segmentation_slice(
    seg_slice,
    x1, x2,
    y1, y2
):

    # Same brain crop used for MRI
    cropped_mask = seg_slice[
        x1:x2+1,
        y1:y2+1
    ]

    # Same square padding
    padded_mask = pad_to_square(
        cropped_mask
    )

    # Resize using nearest-neighbour interpolation
    processed_mask = resize_to_224(
        padded_mask,
        order=0
    )

    # Keep labels as integers
    processed_mask = np.rint(
        processed_mask
    ).astype(np.uint8)

    return (
        cropped_mask,
        padded_mask,
        processed_mask
    )


# ------------------------------------------------------------
# 2. PROCESS TIME1 SEGMENTATION ON SAME SLICE
# ------------------------------------------------------------

seg1_slice = seg1[:, :, best_z]

(
    seg1_crop,
    seg1_pad,
    seg1_processed
) = preprocess_segmentation_slice(
    seg1_slice,
    x1, x2,
    y1, y2
)


# ------------------------------------------------------------
# 3. PROCESS TIME2 SEGMENTATION
# ------------------------------------------------------------

seg2_slice = seg2[:, :, best_z]

(
    seg2_crop,
    seg2_pad,
    seg2_processed
) = preprocess_segmentation_slice(
    seg2_slice,
    x1, x2,
    y1, y2
)


# ------------------------------------------------------------
# 4. CREATE PROCESSED ET MASKS
# ET label = 3
# ------------------------------------------------------------

processed_et1 = (
    seg1_processed == 3
)

processed_et2 = (
    seg2_processed == 3
)


# ------------------------------------------------------------
# 5. PROCESS MATCHING TIME1 + TIME2 T1CE MRI
# ------------------------------------------------------------

t1_crop = n_t1_t1ce[
    x1:x2+1,
    y1:y2+1,
    best_z
]

t2_crop = n_t2_t1ce[
    x1:x2+1,
    y1:y2+1,
    best_z
]

t1_processed = resize_to_224(
    pad_to_square(t1_crop),
    order=1
)

t2_processed = resize_to_224(
    pad_to_square(t2_crop),
    order=1
)


# ------------------------------------------------------------
# 6. VERIFY LABEL VALUES
# ------------------------------------------------------------

print(
    "Original Time1 segmentation labels:",
    np.unique(seg1_slice).astype(int)
)

print(
    "Processed Time1 segmentation labels:",
    np.unique(seg1_processed)
)

print(
    "\nOriginal Time2 segmentation labels:",
    np.unique(seg2_slice).astype(int)
)

print(
    "Processed Time2 segmentation labels:",
    np.unique(seg2_processed)
)

print(
    "\nProcessed segmentation shape:",
    seg1_processed.shape
)


# ------------------------------------------------------------
# 7. VISUALIZE PROCESSED MRI + PROCESSED ET MASK
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2,
    2,
    figsize=(10, 9)
)


# Time1 processed mask
axes[0, 0].imshow(
    processed_et1,
    cmap="gray"
)

axes[0, 0].set_title(
    "Processed Time1 ET Mask"
)

axes[0, 0].axis("off")


# Time2 processed mask
axes[0, 1].imshow(
    processed_et2,
    cmap="gray"
)

axes[0, 1].set_title(
    "Processed Time2 ET Mask"
)

axes[0, 1].axis("off")


# Time1 overlay
axes[1, 0].imshow(
    t1_processed,
    cmap="gray"
)

axes[1, 0].imshow(
    np.ma.masked_where(
        ~processed_et1,
        processed_et1
    ),
    alpha=0.45
)

axes[1, 0].contour(
    processed_et1,
    levels=[0.5],
    linewidths=1.3
)

axes[1, 0].set_title(
    "Processed Time1 T1CE + ET"
)

axes[1, 0].axis("off")


# Time2 overlay
axes[1, 1].imshow(
    t2_processed,
    cmap="gray"
)

axes[1, 1].imshow(
    np.ma.masked_where(
        ~processed_et2,
        processed_et2
    ),
    alpha=0.45
)

axes[1, 1].contour(
    processed_et2,
    levels=[0.5],
    linewidths=1.3
)

axes[1, 1].set_title(
    "Processed Time2 T1CE + ET"
)

axes[1, 1].axis("off")


plt.suptitle(
    f"Patient {patient_id} — Processed MRI / Mask Alignment",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 4 FINAL: PROCESS AND SAVE ALL 298 PATIENTS
#
# Purpose:
#   প্রতিটি patient-এর:
#       Time1 FLAIR + T1CE
#       Time2 FLAIR + T1CE
#       Time1 segmentation
#       Time2 segmentation
#
#   একই:
#       normalization
#       brain crop
#       square padding
#       224x224 resize
#       48 matched axial slices
#
#   দিয়ে preprocess করে compressed .npz হিসেবে save করা।
#
# IMPORTANT:
#   - Tumor mask দিয়ে slice select করা হচ্ছে না
#   - Time1 এবং Time2 একই slice positions ব্যবহার করছে
#   - Segmentation nearest-neighbour interpolation ব্যবহার করছে
#   - Transformation metadata-ও save হচ্ছে
# ============================================================

import os
import numpy as np
import nibabel as nib
from scipy.ndimage import zoom
from tqdm.auto import tqdm


# ------------------------------------------------------------
# FINAL SETTINGS
# ------------------------------------------------------------

IMAGE_SIZE = 224
NUM_SLICES = 48
BRAIN_AREA_THRESHOLD = 0.10

PROCESSED_DIR = os.path.join(
    DATA_DIR,
    "processed_final"
)

os.makedirs(
    PROCESSED_DIR,
    exist_ok=True
)


# ------------------------------------------------------------
# MRI NORMALIZATION
# ------------------------------------------------------------

def normalize_volume(volume):

    volume = volume.astype(np.float32)

    brain = volume != 0

    if brain.sum() == 0:
        return volume

    values = volume[brain]

    low, high = np.percentile(
        values,
        [0.5, 99.5]
    )

    clipped = np.clip(
        volume,
        low,
        high
    )

    mean = clipped[brain].mean()
    std = clipped[brain].std()

    result = np.zeros_like(
        clipped,
        dtype=np.float32
    )

    result[brain] = (
        clipped[brain] - mean
    ) / (std + 1e-6)

    return result


# ------------------------------------------------------------
# SQUARE PADDING
# ------------------------------------------------------------

def pad_square(image):

    h, w = image.shape

    size = max(h, w)

    output = np.zeros(
        (size, size),
        dtype=image.dtype
    )

    top = (size - h) // 2
    left = (size - w) // 2

    output[
        top:top+h,
        left:left+w
    ] = image

    return output, top, left, size


# ------------------------------------------------------------
# RESIZE
# ------------------------------------------------------------

def resize_image(image, order):

    h, w = image.shape

    return zoom(
        image,
        (
            IMAGE_SIZE / h,
            IMAGE_SIZE / w
        ),
        order=order
    )


# ------------------------------------------------------------
# PROCESS ONE PATIENT
# ------------------------------------------------------------

def process_one_patient(pid):

    folder = os.path.join(
        DATA_DIR,
        pid
    )


    # ========================================================
    # 1. LOAD MRI
    # ========================================================

    t1_flair = nib.load(
        os.path.join(
            folder,
            f"{pid}_time1_flair.nii.gz"
        )
    ).get_fdata()

    t1_t1ce = nib.load(
        os.path.join(
            folder,
            f"{pid}_time1_t1ce.nii.gz"
        )
    ).get_fdata()

    t2_flair = nib.load(
        os.path.join(
            folder,
            f"{pid}_time2_flair.nii.gz"
        )
    ).get_fdata()

    t2_t1ce = nib.load(
        os.path.join(
            folder,
            f"{pid}_time2_t1ce.nii.gz"
        )
    ).get_fdata()


    # ========================================================
    # 2. LOAD SEGMENTATIONS
    # ========================================================

    seg1 = nib.load(
        os.path.join(
            folder,
            f"{pid}_time1_seg.nii.gz"
        )
    ).get_fdata()

    seg2 = nib.load(
        os.path.join(
            folder,
            f"{pid}_time2_seg.nii.gz"
        )
    ).get_fdata()


    original_shape = np.array(
        t1_flair.shape,
        dtype=np.int16
    )


    # ========================================================
    # 3. COMMON BRAIN MASK
    # ========================================================

    brain_union = (
        (t1_flair != 0) |
        (t1_t1ce != 0) |
        (t2_flair != 0) |
        (t2_t1ce != 0)
    )


    coords = np.argwhere(
        brain_union
    )


    # ========================================================
    # 4. COMMON XY BOUNDING BOX
    # ========================================================

    mins = coords.min(axis=0)
    maxs = coords.max(axis=0)

    padding = 5

    x1 = max(
        0,
        int(mins[0]) - padding
    )

    x2 = min(
        t1_flair.shape[0] - 1,
        int(maxs[0]) + padding
    )

    y1 = max(
        0,
        int(mins[1]) - padding
    )

    y2 = min(
        t1_flair.shape[1] - 1,
        int(maxs[1]) + padding
    )


    # ========================================================
    # 5. CORE-BRAIN SLICE SELECTION
    # ========================================================

    brain_area = brain_union.sum(
        axis=(0, 1)
    )

    max_area = brain_area.max()

    core_slices = np.where(
        brain_area >=
        BRAIN_AREA_THRESHOLD * max_area
    )[0]


    # Pick 48 positions from actual core-brain slices
    positions = np.linspace(
        0,
        len(core_slices) - 1,
        NUM_SLICES
    )

    positions = np.round(
        positions
    ).astype(int)

    selected_z = core_slices[
        positions
    ]


    # ========================================================
    # 6. NORMALIZE MRI VOLUMES
    # ========================================================

    t1_flair = normalize_volume(
        t1_flair
    )

    t1_t1ce = normalize_volume(
        t1_t1ce
    )

    t2_flair = normalize_volume(
        t2_flair
    )

    t2_t1ce = normalize_volume(
        t2_t1ce
    )


    # ========================================================
    # 7. OUTPUT ARRAYS
    # ========================================================

    time1 = []
    time2 = []

    masks1 = []
    masks2 = []


    pad_top = None
    pad_left = None
    square_size = None


    # ========================================================
    # 8. PROCESS 48 MATCHED SLICES
    # ========================================================

    for z in selected_z:

        # ------------------------
        # MRI crop
        # ------------------------

        a = t1_flair[
            x1:x2+1,
            y1:y2+1,
            z
        ]

        b = t1_t1ce[
            x1:x2+1,
            y1:y2+1,
            z
        ]

        c = t2_flair[
            x1:x2+1,
            y1:y2+1,
            z
        ]

        d = t2_t1ce[
            x1:x2+1,
            y1:y2+1,
            z
        ]


        # ------------------------
        # Segmentation crop
        # ------------------------

        m1 = seg1[
            x1:x2+1,
            y1:y2+1,
            z
        ]

        m2 = seg2[
            x1:x2+1,
            y1:y2+1,
            z
        ]


        # ------------------------
        # Same square padding
        # ------------------------

        a, pt, pl, ss = pad_square(a)
        b, _, _, _ = pad_square(b)

        c, _, _, _ = pad_square(c)
        d, _, _, _ = pad_square(d)

        m1, _, _, _ = pad_square(m1)
        m2, _, _, _ = pad_square(m2)


        pad_top = pt
        pad_left = pl
        square_size = ss


        # ------------------------
        # Resize MRI
        # ------------------------

        a = resize_image(a, order=1)
        b = resize_image(b, order=1)

        c = resize_image(c, order=1)
        d = resize_image(d, order=1)


        # ------------------------
        # Resize masks
        # nearest-neighbour
        # ------------------------

        m1 = resize_image(
            m1,
            order=0
        )

        m2 = resize_image(
            m2,
            order=0
        )


        # ------------------------
        # Two MRI channels
        # ------------------------

        time1.append(
            np.stack([a, b])
        )

        time2.append(
            np.stack([c, d])
        )


        masks1.append(
            np.rint(m1)
        )

        masks2.append(
            np.rint(m2)
        )


    # ========================================================
    # 9. COMPACT DATA TYPES
    # ========================================================

    time1 = np.asarray(
        time1,
        dtype=np.float16
    )

    time2 = np.asarray(
        time2,
        dtype=np.float16
    )

    masks1 = np.asarray(
        masks1,
        dtype=np.uint8
    )

    masks2 = np.asarray(
        masks2,
        dtype=np.uint8
    )


    # ========================================================
    # 10. SAVE
    # ========================================================

    save_path = os.path.join(
        PROCESSED_DIR,
        f"{pid}.npz"
    )


    np.savez_compressed(
        save_path,

        time1=time1,
        time2=time2,

        seg1=masks1,
        seg2=masks2,

        slice_ids=selected_z.astype(
            np.int16
        ),

        original_shape=original_shape,

        bbox=np.array(
            [x1, x2, y1, y2],
            dtype=np.int16
        ),

        pad_top=np.int16(pad_top),
        pad_left=np.int16(pad_left),

        square_size=np.int16(
            square_size
        )
    )


# ============================================================
# 11. PROCESS ALL 298 PATIENTS
# ============================================================

failed = []


for pid in tqdm(
    patient_ids,
    desc="Processing 298 patients"
):

    save_path = os.path.join(
        PROCESSED_DIR,
        f"{pid}.npz"
    )


    # Resume-safe
    if os.path.exists(save_path):
        continue


    try:

        process_one_patient(pid)

    except Exception as e:

        failed.append(
            (pid, str(e))
        )


# ============================================================
# 12. FINAL CHECK
# ============================================================

processed_files = [
    f for f in os.listdir(
        PROCESSED_DIR
    )
    if f.endswith(".npz")
]


print("\n==============================")
print("PROCESSING COMPLETE")
print("==============================")

print(
    "Processed patients:",
    len(processed_files)
)

print(
    "Failed patients:",
    len(failed)
)


if failed:

    print("\nFailed list:")

    for item in failed[:10]:
        print(item)
```

```python
# ============================================================
# STEP 6: STRATIFIED 70/15/15 PATIENT-LEVEL SPLIT
#
# Purpose:
#   298 patients কে Train / Validation / Test সেটে ভাগ করা
#
# Split:
#   Train      = 70%
#   Validation = 15%
#   Test       = 15%
#
# কেন stratified?
#   Unchanged / Increased / Decreased class imbalance আছে।
#   তাই প্রতিটি split-এ class proportion যতটা সম্ভব একই রাখা।
#
# IMPORTANT:
#   Test set এখন থেকে final evaluation-এর আগে touch করা হবে না.
# ============================================================

import os
import pandas as pd
from sklearn.model_selection import train_test_split


SEED = 42


# ------------------------------------------------------------
# 1. First split:
#    70% Train
#    30% Temporary
# ------------------------------------------------------------

train_df, temp_df = train_test_split(
    clinical_df,
    test_size=0.30,
    stratify=clinical_df["Label"],
    random_state=SEED
)


# ------------------------------------------------------------
# 2. Split temporary data equally:
#    15% Validation
#    15% Test
# ------------------------------------------------------------

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["Label"],
    random_state=SEED
)


# ------------------------------------------------------------
# 3. Add permanent split name
# ------------------------------------------------------------

train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df["Split"] = "train"
val_df["Split"] = "val"
test_df["Split"] = "test"


# ------------------------------------------------------------
# 4. Combine into one master metadata table
# ------------------------------------------------------------

split_df = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)


# ------------------------------------------------------------
# 5. Save permanently to Google Drive
# ------------------------------------------------------------

METADATA_DIR = os.path.join(
    DATA_DIR,
    "research_metadata"
)

os.makedirs(
    METADATA_DIR,
    exist_ok=True
)

split_path = os.path.join(
    METADATA_DIR,
    "patient_split.csv"
)

split_df.to_csv(
    split_path,
    index=False
)


# ------------------------------------------------------------
# 6. Print results
# ------------------------------------------------------------

print("===== SPLIT SIZE =====")

print("Train      :", len(train_df))
print("Validation :", len(val_df))
print("Test       :", len(test_df))


print("\n===== TRAIN DISTRIBUTION =====")
print(train_df["Label"].value_counts())


print("\n===== VALIDATION DISTRIBUTION =====")
print(val_df["Label"].value_counts())


print("\n===== TEST DISTRIBUTION =====")
print(test_df["Label"].value_counts())


print("\nSaved to:")
print(split_path)


# ------------------------------------------------------------
# 7. Safety checks
# ------------------------------------------------------------

assert len(split_df) == 298

assert set(train_df["PatientID"]).isdisjoint(
    set(val_df["PatientID"])
)

assert set(train_df["PatientID"]).isdisjoint(
    set(test_df["PatientID"])
)

assert set(val_df["PatientID"]).isdisjoint(
    set(test_df["PatientID"])
)

print("\n✅ Patient-level split verification PASSED.")
```

```python
# ============================================================
# STEP 7: CLINICAL PREPROCESSING + CLASS IMBALANCE HANDLING
#
# Purpose:
#   1. Age standardize করা
#   2. Sex + Diagnosis encode করা
#   3. Transformer ONLY training set-এ fit করা
#   4. Train/Val/Test clinical features transform করা
#   5. Training labels থেকে class weights calculate করা
#
# IMPORTANT:
#   Validation/Test data দিয়ে preprocessing fit করা হচ্ছে না.
#   এতে data leakage এড়ানো হয়.
# ============================================================

import os
import json
import joblib
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.utils.class_weight import compute_class_weight


# ------------------------------------------------------------
# 1. Feature columns
# ------------------------------------------------------------

numeric_features = ["Age"]

categorical_features = [
    "Sex",
    "Diagnosis"
]


# ------------------------------------------------------------
# 2. Build preprocessing pipeline
# ------------------------------------------------------------

clinical_preprocessor = ColumnTransformer(
    transformers=[
        (
            "age",
            StandardScaler(),
            numeric_features
        ),

        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)


# ------------------------------------------------------------
# 3. Fit ONLY on training patients
# ------------------------------------------------------------

X_train_clinical = clinical_preprocessor.fit_transform(
    train_df[
        numeric_features + categorical_features
    ]
)


# ------------------------------------------------------------
# 4. Transform validation and test
# ------------------------------------------------------------

X_val_clinical = clinical_preprocessor.transform(
    val_df[
        numeric_features + categorical_features
    ]
)

X_test_clinical = clinical_preprocessor.transform(
    test_df[
        numeric_features + categorical_features
    ]
)


# ------------------------------------------------------------
# 5. Label mapping
# ------------------------------------------------------------

label_map = {
    "Unchanged": 0,
    "Increased": 1,
    "Decreased": 2
}


y_train = (
    train_df["Label"]
    .map(label_map)
    .to_numpy()
)

y_val = (
    val_df["Label"]
    .map(label_map)
    .to_numpy()
)

y_test = (
    test_df["Label"]
    .map(label_map)
    .to_numpy()
)


# ------------------------------------------------------------
# 6. Calculate class weights from TRAINING SET ONLY
# ------------------------------------------------------------

classes = np.array([0, 1, 2])

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = class_weights.astype(
    np.float32
)


# ------------------------------------------------------------
# 7. Save everything permanently
# ------------------------------------------------------------

joblib.dump(
    clinical_preprocessor,
    os.path.join(
        METADATA_DIR,
        "clinical_preprocessor.joblib"
    )
)


np.savez_compressed(
    os.path.join(
        METADATA_DIR,
        "clinical_features.npz"
    ),

    X_train=X_train_clinical.astype(np.float32),
    X_val=X_val_clinical.astype(np.float32),
    X_test=X_test_clinical.astype(np.float32),

    y_train=y_train,
    y_val=y_val,
    y_test=y_test
)


np.save(
    os.path.join(
        METADATA_DIR,
        "class_weights.npy"
    ),
    class_weights
)


with open(
    os.path.join(
        METADATA_DIR,
        "label_map.json"
    ),
    "w"
) as f:

    json.dump(
        label_map,
        f,
        indent=4
    )


# ------------------------------------------------------------
# 8. Show results
# ------------------------------------------------------------

print("===== CLINICAL FEATURES =====")

print(
    "Train shape:",
    X_train_clinical.shape
)

print(
    "Validation shape:",
    X_val_clinical.shape
)

print(
    "Test shape:",
    X_test_clinical.shape
)


print(
    "\nClinical feature dimension:",
    X_train_clinical.shape[1]
)


print("\n===== CLASS WEIGHTS =====")

for name, idx in label_map.items():

    print(
        f"{name:10s}:",
        round(
            float(class_weights[idx]),
            4
        )
    )


print(
    "\n✅ Clinical preprocessing saved."
)

print(
    "✅ Class imbalance weights saved."
)
```

```python
# ============================================================
# STEP 8: CLINICAL-ONLY BASELINE
#
# Purpose:
#   শুধুমাত্র clinical features:
#       Age
#       Sex
#       Diagnosis
#
#   ব্যবহার করে progression predict করা।
#
# Model:
#   Balanced Logistic Regression
#
# IMPORTANT:
#   Final TEST SET এখন ব্যবহার করছি না.
#   শুধু TRAIN দিয়ে fit এবং VALIDATION দিয়ে evaluate করছি.
# ============================================================

import os
import joblib
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)


# ------------------------------------------------------------
# 1. Create class-weight dictionary
# ------------------------------------------------------------

weight_dict = {
    0: float(class_weights[0]),
    1: float(class_weights[1]),
    2: float(class_weights[2])
}


# ------------------------------------------------------------
# 2. Create clinical-only baseline model
# ------------------------------------------------------------

clinical_model = LogisticRegression(
    class_weight=weight_dict,
    max_iter=2000,
    random_state=SEED
)


# ------------------------------------------------------------
# 3. Train using TRAIN set only
# ------------------------------------------------------------

clinical_model.fit(
    X_train_clinical,
    y_train
)


# ------------------------------------------------------------
# 4. Validation prediction
# ------------------------------------------------------------

val_pred = clinical_model.predict(
    X_val_clinical
)


# ------------------------------------------------------------
# 5. Metrics
# ------------------------------------------------------------

val_accuracy = accuracy_score(
    y_val,
    val_pred
)

val_macro_f1 = f1_score(
    y_val,
    val_pred,
    average="macro"
)


print("===== CLINICAL-ONLY VALIDATION =====")

print(
    f"Validation Accuracy : {val_accuracy:.4f}"
)

print(
    f"Validation Macro-F1 : {val_macro_f1:.4f}"
)


print("\n===== CLASSIFICATION REPORT =====")

print(
    classification_report(
        y_val,
        val_pred,
        target_names=[
            "Unchanged",
            "Increased",
            "Decreased"
        ],
        digits=4,
        zero_division=0
    )
)


# ------------------------------------------------------------
# 6. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_val,
    val_pred,
    labels=[0, 1, 2]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Unchanged",
        "Increased",
        "Decreased"
    ]
)

disp.plot()

plt.title(
    "Clinical-Only Baseline — Validation"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. Save baseline model
# ------------------------------------------------------------

MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

clinical_model_path = os.path.join(
    MODEL_DIR,
    "clinical_only_logistic.joblib"
)

joblib.dump(
    clinical_model,
    clinical_model_path
)

print(
    "\n✅ Clinical-only baseline saved:"
)

print(
    clinical_model_path
)
```

```python
# ============================================================
# STEP 9.1: TIME1 MRI DATASET + DATALOADER
#
# Purpose:
#   processed_final/*.npz থেকে Time1 MRI load করা।
#
# Each patient:
#   48 slices
#   2 channels = FLAIR + T1CE
#
# Output tensor:
#   (48, 2, 224, 224)
#
# IMPORTANT:
#   Test set এখন ব্যবহার করা হচ্ছে না.
# ============================================================

import os
import numpy as np
import torch

from torch.utils.data import Dataset, DataLoader


# ------------------------------------------------------------
# Label mapping
# ------------------------------------------------------------

label_map = {
    "Unchanged": 0,
    "Increased": 1,
    "Decreased": 2
}


# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------

class Time1MRIDataset(Dataset):

    def __init__(self, dataframe, processed_dir):

        self.df = dataframe.reset_index(drop=True)
        self.processed_dir = processed_dir


    def __len__(self):

        return len(self.df)


    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        patient_id = row["PatientID"]

        file_path = os.path.join(
            self.processed_dir,
            f"{patient_id}.npz"
        )

        data = np.load(file_path)

        # Shape:
        # (48, 2, 224, 224)
        images = data["time1"].astype(
            np.float32
        )

        label = label_map[
            row["Label"]
        ]


        images = torch.from_numpy(
            images
        )

        label = torch.tensor(
            label,
            dtype=torch.long
        )

        return images, label


# ------------------------------------------------------------
# Create datasets
# ------------------------------------------------------------

train_dataset_time1 = Time1MRIDataset(
    train_df,
    PROCESSED_DIR
)

val_dataset_time1 = Time1MRIDataset(
    val_df,
    PROCESSED_DIR
)


# ------------------------------------------------------------
# DataLoaders
#
# Batch size 1 রাখা হচ্ছে কারণ:
# 48 large 224×224 MRI slices এক patient-এ আছে.
# ------------------------------------------------------------

train_loader_time1 = DataLoader(
    train_dataset_time1,
    batch_size=1,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader_time1 = DataLoader(
    val_dataset_time1,
    batch_size=1,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ------------------------------------------------------------
# Check one batch
# ------------------------------------------------------------

images, labels = next(
    iter(train_loader_time1)
)

print(
    "MRI batch shape:",
    images.shape
)

print(
    "Label shape:",
    labels.shape
)

print(
    "Training patients:",
    len(train_dataset_time1)
)

print(
    "Validation patients:",
    len(val_dataset_time1)
)
```

```python
# ============================================================
# STEP 9.2: SINGLE-TIME1 PRETRAINED CNN MODEL
#
# Purpose:
#   প্রতি patient-এর 48টি Time1 MRI slice থেকে
#   pretrained ResNet18 দিয়ে feature extract করা।
#
# Input:
#   [Batch, 48 slices, 2 channels, 224, 224]
#
# Channels:
#   0 = FLAIR
#   1 = T1CE
#
# Output:
#   3 progression classes:
#   0 = Unchanged
#   1 = Increased
#   2 = Decreased
# ============================================================

import torch
import torch.nn as nn

from torchvision.models import (
    resnet18,
    ResNet18_Weights
)


class Time1PatientCNN(nn.Module):

    def __init__(self, num_classes=3):

        super().__init__()

        # ----------------------------------------------------
        # 1. Load pretrained ResNet18
        # ----------------------------------------------------

        backbone = resnet18(
            weights=ResNet18_Weights.DEFAULT
        )


        # ----------------------------------------------------
        # 2. Convert input from RGB 3-channel
        #    to MRI 2-channel (FLAIR + T1CE)
        # ----------------------------------------------------

        original_conv = backbone.conv1

        backbone.conv1 = nn.Conv2d(
            in_channels=2,
            out_channels=original_conv.out_channels,
            kernel_size=original_conv.kernel_size,
            stride=original_conv.stride,
            padding=original_conv.padding,
            bias=False
        )


        # ----------------------------------------------------
        # 3. Initialize new 2-channel layer
        #    using pretrained RGB weights
        # ----------------------------------------------------

        with torch.no_grad():

            mean_weight = original_conv.weight.mean(
                dim=1,
                keepdim=True
            )

            backbone.conv1.weight.copy_(
                mean_weight.repeat(1, 2, 1, 1)
            )


        # ----------------------------------------------------
        # 4. Remove original ImageNet classifier
        # ----------------------------------------------------

        feature_dim = backbone.fc.in_features

        backbone.fc = nn.Identity()

        self.encoder = backbone


        # ----------------------------------------------------
        # 5. Patient-level classifier
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Linear(
                feature_dim,
                128
            ),

            nn.ReLU(),

            nn.Dropout(0.4),

            nn.Linear(
                128,
                num_classes
            )
        )


    def forward(self, x):

        # x:
        # [B, 48, 2, 224, 224]

        batch_size, slices, channels, height, width = x.shape


        # ----------------------------------------------------
        # Treat MRI slices as CNN images
        # ----------------------------------------------------

        x = x.reshape(
            batch_size * slices,
            channels,
            height,
            width
        )


        # ----------------------------------------------------
        # Extract feature for each slice
        # ----------------------------------------------------

        slice_features = self.encoder(x)

        # shape:
        # [B*48, 512]


        # ----------------------------------------------------
        # Restore patient grouping
        # ----------------------------------------------------

        slice_features = slice_features.reshape(
            batch_size,
            slices,
            -1
        )


        # ----------------------------------------------------
        # Aggregate 48 slices into one patient representation
        # ----------------------------------------------------

        patient_feature = slice_features.mean(
            dim=1
        )


        # ----------------------------------------------------
        # Progression prediction
        # ----------------------------------------------------

        logits = self.classifier(
            patient_feature
        )

        return logits


# ============================================================
# CREATE MODEL ON GPU
# ============================================================

model_time1 = Time1PatientCNN(
    num_classes=3
).to(device)


print("✅ Time1 pretrained CNN created.")
print("Device:", device)


# ============================================================
# SANITY CHECK WITH ONE PATIENT
# ============================================================

model_time1.eval()

with torch.no_grad():

    sample_images, _ = next(
        iter(train_loader_time1)
    )

    sample_images = sample_images.to(
        device
    )

    sample_output = model_time1(
        sample_images
    )


print(
    "Input shape :",
    sample_images.shape
)

print(
    "Output shape:",
    sample_output.shape
)
```

```python
# ============================================================
# STEP 9.3: ACTUAL TIME1 MRI CNN TRAINING
#
# Purpose:
#   Time1 FLAIR + T1CE MRI ব্যবহার করে CNN train করা।
#
# Features:
#   - Pretrained ResNet18
#   - Class-weighted loss
#   - Mixed Precision (T4 GPU faster)
#   - AdamW optimizer
#   - Learning-rate scheduler
#   - Validation Macro-F1 monitoring
#   - Early stopping
#   - Best model automatically saved to Google Drive
#
# IMPORTANT:
#   Test set এখানে ব্যবহার করা হচ্ছে না।
# ============================================================

import os
import numpy as np
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    f1_score
)


# ------------------------------------------------------------
# 1. Freeze early ResNet layers
#
# Small dataset (208 training patients), তাই পুরো pretrained
# network প্রথম থেকেই aggressively train না করে early
# ImageNet features preserve করছি.
# ------------------------------------------------------------

for param in model_time1.encoder.parameters():
    param.requires_grad = False

# Last ResNet block fine-tune করবো
for param in model_time1.encoder.layer4.parameters():
    param.requires_grad = True

# Classifier অবশ্যই train হবে
for param in model_time1.classifier.parameters():
    param.requires_grad = True


# ------------------------------------------------------------
# 2. Weighted Cross Entropy
#
# Rare "Decreased" class-কে বেশি importance দেয়.
# ------------------------------------------------------------

weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=weights_tensor
)


# ------------------------------------------------------------
# 3. Optimizer
# ------------------------------------------------------------

trainable_parameters = [
    p for p in model_time1.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_parameters,
    lr=1e-4,
    weight_decay=1e-4
)


# ------------------------------------------------------------
# 4. Learning-rate scheduler
#
# Validation Macro-F1 improve না করলে LR কমাবে.
# ------------------------------------------------------------

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)


# ------------------------------------------------------------
# 5. Mixed Precision
# T4 GPU training দ্রুত এবং memory-efficient করবে.
# ------------------------------------------------------------

scaler = torch.amp.GradScaler(
    "cuda"
)


# ------------------------------------------------------------
# 6. Training settings
# ------------------------------------------------------------

EPOCHS = 15
EARLY_STOPPING_PATIENCE = 5

best_macro_f1 = -1.0
epochs_without_improvement = 0


# ------------------------------------------------------------
# 7. Model save location
# ------------------------------------------------------------

MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

best_model_path = os.path.join(
    MODEL_DIR,
    "best_time1_cnn.pt"
)


# ============================================================
# 8. TRAINING LOOP
# ============================================================

for epoch in range(EPOCHS):

    # ========================================================
    # TRAINING PHASE
    # ========================================================

    model_time1.train()

    running_loss = 0.0

    train_true = []
    train_pred = []


    for images, labels in train_loader_time1:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        # ----------------------------------------------------
        # Mixed-precision forward pass
        # ----------------------------------------------------

        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_time1(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        # ----------------------------------------------------
        # Backpropagation
        # ----------------------------------------------------

        scaler.scale(
            loss
        ).backward()

        scaler.step(
            optimizer
        )

        scaler.update()


        running_loss += loss.item()


        predictions = outputs.argmax(
            dim=1
        )


        train_true.extend(
            labels.detach().cpu().numpy()
        )

        train_pred.extend(
            predictions.detach().cpu().numpy()
        )


    # --------------------------------------------------------
    # Training metrics
    # --------------------------------------------------------

    train_loss = (
        running_loss /
        len(train_loader_time1)
    )

    train_accuracy = accuracy_score(
        train_true,
        train_pred
    )


    # ========================================================
    # VALIDATION PHASE
    # ========================================================

    model_time1.eval()

    val_true = []
    val_pred = []

    val_running_loss = 0.0


    with torch.no_grad():

        for images, labels in val_loader_time1:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model_time1(
                    images
                )

                loss = criterion(
                    outputs,
                    labels
                )


            val_running_loss += loss.item()


            predictions = outputs.argmax(
                dim=1
            )


            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                predictions.cpu().numpy()
            )


    # --------------------------------------------------------
    # Validation metrics
    # --------------------------------------------------------

    val_loss = (
        val_running_loss /
        len(val_loader_time1)
    )

    val_accuracy = accuracy_score(
        val_true,
        val_pred
    )

    val_macro_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # Learning-rate update
    # --------------------------------------------------------

    scheduler.step(
        val_macro_f1
    )


    current_lr = optimizer.param_groups[0]["lr"]


    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    if val_macro_f1 > best_macro_f1:

        best_macro_f1 = val_macro_f1

        epochs_without_improvement = 0


        torch.save(
            {
                "epoch": epoch + 1,

                "model_state_dict":
                    model_time1.state_dict(),

                "optimizer_state_dict":
                    optimizer.state_dict(),

                "val_macro_f1":
                    val_macro_f1,

                "val_accuracy":
                    val_accuracy
            },

            best_model_path
        )


        best_status = "  ⭐ BEST SAVED"


    else:

        epochs_without_improvement += 1

        best_status = ""


    # ========================================================
    # PRINT EPOCH RESULT
    # ========================================================

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Val Macro-F1: {val_macro_f1:.4f} | "
        f"LR: {current_lr:.6f}"
        f"{best_status}"
    )


    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

        print(
            "\nEarly stopping triggered."
        )

        break


# ============================================================
# FINAL RESULT
# ============================================================

print("\n======================================")
print("TIME1 MRI TRAINING COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_macro_f1, 4)
)

print(
    "Best model saved at:"
)

print(
    best_model_path
)
```

```python
# ============================================================
# STEP 11.1: SIAMESE PAIRED MRI DATALOADER
#
# Purpose:
#   একই patient-এর Time1 এবং Time2 MRI একসাথে load করা
#   যাতে True Siamese CNN longitudinal change শিখতে পারে।
#
# Each timepoint:
#   48 slices
#   2 channels = FLAIR + T1CE
#   224 x 224
#
# Output:
#   Time1 tensor + Time2 tensor + progression label
# ============================================================

import os
import numpy as np
import torch

from torch.utils.data import Dataset, DataLoader


label_map = {
    "Unchanged": 0,
    "Increased": 1,
    "Decreased": 2
}


class SiameseMRIDataset(Dataset):

    def __init__(self, dataframe, processed_dir):

        self.df = dataframe.reset_index(drop=True)
        self.processed_dir = processed_dir


    def __len__(self):

        return len(self.df)


    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        patient_id = str(
            row["PatientID"]
        )

        file_path = os.path.join(
            self.processed_dir,
            f"{patient_id}.npz"
        )


        # --------------------------------
        # Load paired longitudinal MRI
        # --------------------------------

        with np.load(file_path) as data:

            time1 = data["time1"].astype(
                np.float32
            )

            time2 = data["time2"].astype(
                np.float32
            )


        label = label_map[
            row["Label"]
        ]


        time1 = torch.from_numpy(
            time1
        )

        time2 = torch.from_numpy(
            time2
        )

        label = torch.tensor(
            label,
            dtype=torch.long
        )


        return time1, time2, label


# ============================================================
# CREATE TRAIN / VALIDATION DATASETS
# ============================================================

train_siamese_dataset = SiameseMRIDataset(
    train_df,
    PROCESSED_DIR
)

val_siamese_dataset = SiameseMRIDataset(
    val_df,
    PROCESSED_DIR
)


# ============================================================
# DATALOADERS
# ============================================================

train_siamese_loader = DataLoader(
    train_siamese_dataset,
    batch_size=1,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_siamese_loader = DataLoader(
    val_siamese_dataset,
    batch_size=1,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# SANITY CHECK
# ============================================================

time1_batch, time2_batch, label_batch = next(
    iter(train_siamese_loader)
)


print(
    "Time1 shape:",
    time1_batch.shape
)

print(
    "Time2 shape:",
    time2_batch.shape
)

print(
    "Label shape:",
    label_batch.shape
)

print(
    "Training patients:",
    len(train_siamese_dataset)
)

print(
    "Validation patients:",
    len(val_siamese_dataset)
)

print(
    "\n✅ Siamese paired DataLoader ready."
)
```

```python
# ============================================================
# STEP 11.2: TRUE SIAMESE CNN ARCHITECTURE
#
# Purpose:
#   Time1 MRI -> Shared Encoder -> F1
#   Time2 MRI -> SAME Encoder   -> F2
#
#   Longitudinal change:
#       Delta_F = F2 - F1
#
#   F1 + F2 + Delta_F combine করে progression predict করা।
#
# IMPORTANT:
#   দুই branch-এর আলাদা CNN নেই।
#   SAME encoder দুই timepoint-এ ব্যবহার হচ্ছে।
#
# Output:
#   0 = Unchanged
#   1 = Increased
#   2 = Decreased
# ============================================================

import torch
import torch.nn as nn

from torchvision.models import (
    resnet18,
    ResNet18_Weights
)


class SiameseGliomaCNN(nn.Module):

    def __init__(self, num_classes=3):

        super().__init__()

        # ----------------------------------------------------
        # 1. Pretrained ResNet18 shared encoder
        # ----------------------------------------------------

        backbone = resnet18(
            weights=ResNet18_Weights.DEFAULT
        )


        # ----------------------------------------------------
        # 2. RGB 3-channel -> MRI 2-channel
        #    FLAIR + T1CE
        # ----------------------------------------------------

        old_conv = backbone.conv1

        backbone.conv1 = nn.Conv2d(
            in_channels=2,
            out_channels=old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding,
            bias=False
        )


        # ----------------------------------------------------
        # 3. Initialize 2-channel layer from pretrained weights
        # ----------------------------------------------------

        with torch.no_grad():

            mean_weight = old_conv.weight.mean(
                dim=1,
                keepdim=True
            )

            backbone.conv1.weight.copy_(
                mean_weight.repeat(1, 2, 1, 1)
            )


        # ResNet feature dimension = 512
        self.feature_dim = backbone.fc.in_features

        # Remove ImageNet classifier
        backbone.fc = nn.Identity()

        # ONE shared encoder
        self.encoder = backbone


        # ----------------------------------------------------
        # 4. Longitudinal feature fusion
        #
        # F1
        # F2
        # F2 - F1
        #
        # 512 × 3 = 1536
        # ----------------------------------------------------

        fusion_dim = self.feature_dim * 3


        # ----------------------------------------------------
        # 5. Progression classifier
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Linear(
                fusion_dim,
                256
            ),

            nn.BatchNorm1d(256),

            nn.ReLU(),

            nn.Dropout(0.4),

            nn.Linear(
                256,
                64
            ),

            nn.ReLU(),

            nn.Dropout(0.3),

            nn.Linear(
                64,
                num_classes
            )
        )


    # ========================================================
    # Patient-level MRI encoding
    # ========================================================

    def encode_patient(self, x):

        # x:
        # [Batch, 48, 2, 224, 224]

        B, S, C, H, W = x.shape


        # 48 slices -> CNN images
        x = x.reshape(
            B * S,
            C,
            H,
            W
        )


        # Shared ResNet features
        slice_features = self.encoder(x)

        # [B*48, 512]
        slice_features = slice_features.reshape(
            B,
            S,
            self.feature_dim
        )


        # ----------------------------------------------------
        # Aggregate 48 slices -> one patient representation
        # ----------------------------------------------------

        patient_feature = slice_features.mean(
            dim=1
        )

        return patient_feature


    # ========================================================
    # Siamese forward pass
    # ========================================================

    def forward(self, time1, time2):

        # SAME encoder
        F1 = self.encode_patient(time1)

        F2 = self.encode_patient(time2)


        # ----------------------------------------------------
        # Signed longitudinal change
        #
        # Positive/negative direction preserve করা হচ্ছে
        # ----------------------------------------------------

        delta_F = F2 - F1


        # ----------------------------------------------------
        # Final longitudinal representation
        # ----------------------------------------------------

        longitudinal_feature = torch.cat(
            [
                F1,
                F2,
                delta_F
            ],
            dim=1
        )


        logits = self.classifier(
            longitudinal_feature
        )


        return logits


# ============================================================
# CREATE MODEL
# ============================================================

model_siamese = SiameseGliomaCNN(
    num_classes=3
).to(device)


print("✅ True Siamese CNN created.")
print("Device:", device)


# ============================================================
# SANITY CHECK
# ============================================================

model_siamese.eval()

with torch.no_grad():

    t1, t2, _ = next(
        iter(train_siamese_loader)
    )

    t1 = t1.to(device)
    t2 = t2.to(device)

    output = model_siamese(
        t1,
        t2
    )


print("Time1 input :", t1.shape)
print("Time2 input :", t2.shape)
print("Model output:", output.shape)
```

```python
# ============================================================
# STEP 11.3: TRUE SIAMESE CNN ACTUAL TRAINING
#
# Purpose:
#   Time1 + Time2 MRI ব্যবহার করে TRUE Siamese CNN train করা।
#
# Training features:
#   - Shared pretrained ResNet18
#   - F1, F2, F2-F1 longitudinal representation
#   - Class-weighted Cross Entropy
#   - Paired MRI augmentation
#   - Mixed precision on T4 GPU
#   - LR scheduler
#   - Early stopping
#   - Best model selected by Validation Macro-F1
#   - Training history saved for thesis plots
#
# Test set এখানে ব্যবহার করা হচ্ছে না.
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    f1_score
)


# ============================================================
# 1. IMPORTANT SMALL-BATCH FIX
# Batch size = 1, তাই BatchNorm1d -> LayerNorm
# ============================================================

model_siamese.classifier[1] = nn.LayerNorm(
    256
).to(device)


# ============================================================
# 2. Freeze early encoder layers
# Small dataset হওয়ায় pretrained knowledge preserve করছি
# ============================================================

for param in model_siamese.encoder.parameters():
    param.requires_grad = False

# Fine-tune last ResNet block
for param in model_siamese.encoder.layer4.parameters():
    param.requires_grad = True

# New 2-channel input layer trainable রাখবো
for param in model_siamese.encoder.conv1.parameters():
    param.requires_grad = True

# Classifier trainable
for param in model_siamese.classifier.parameters():
    param.requires_grad = True


# ============================================================
# 3. CLASS-WEIGHTED LOSS
# ============================================================

weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=weights_tensor
)


# ============================================================
# 4. OPTIMIZER
# ============================================================

trainable_params = [
    p for p in model_siamese.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=1e-4,
    weight_decay=1e-4
)


# ============================================================
# 5. LR SCHEDULER
# ============================================================

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)


# ============================================================
# 6. MIXED PRECISION
# ============================================================

scaler = torch.amp.GradScaler(
    "cuda"
)


# ============================================================
# 7. PAIRED AUGMENTATION
#
# Same geometric augmentation Time1 এবং Time2-তে apply হবে,
# যাতে longitudinal correspondence নষ্ট না হয়.
# ============================================================

def paired_augmentation(time1, time2):

    # Same left-right flip for both timepoints
    if torch.rand(1).item() < 0.5:

        time1 = torch.flip(
            time1,
            dims=[-1]
        )

        time2 = torch.flip(
            time2,
            dims=[-1]
        )


    # Small common intensity variation
    scale = (
        1.0 +
        torch.empty(
            1,
            device=time1.device
        ).uniform_(-0.05, 0.05)
    )

    time1 = time1 * scale
    time2 = time2 * scale

    return time1, time2


# ============================================================
# 8. TRAINING SETTINGS
# ============================================================

EPOCHS = 20
EARLY_STOPPING_PATIENCE = 5

best_macro_f1 = -1.0
epochs_without_improvement = 0

history = []


# ============================================================
# 9. SAVE PATH
# ============================================================

MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

best_siamese_path = os.path.join(
    MODEL_DIR,
    "best_siamese_mri_only.pt"
)

history_path = os.path.join(
    MODEL_DIR,
    "siamese_mri_training_history.csv"
)


# ============================================================
# 10. TRAINING LOOP
# ============================================================

for epoch in range(EPOCHS):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model_siamese.train()

    train_losses = []
    train_true = []
    train_pred = []


    for time1, time2, labels in train_siamese_loader:

        time1 = time1.to(
            device,
            non_blocking=True
        )

        time2 = time2.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        # Paired augmentation
        time1, time2 = paired_augmentation(
            time1,
            time2
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_siamese(
                time1,
                time2
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(
            loss
        ).backward()

        scaler.step(
            optimizer
        )

        scaler.update()


        train_losses.append(
            loss.item()
        )


        predictions = outputs.argmax(
            dim=1
        )


        train_true.extend(
            labels.detach().cpu().numpy()
        )

        train_pred.extend(
            predictions.detach().cpu().numpy()
        )


    train_loss = np.mean(
        train_losses
    )

    train_acc = accuracy_score(
        train_true,
        train_pred
    )

    train_macro_f1 = f1_score(
        train_true,
        train_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model_siamese.eval()

    val_losses = []
    val_true = []
    val_pred = []


    with torch.no_grad():

        for time1, time2, labels in val_siamese_loader:

            time1 = time1.to(
                device,
                non_blocking=True
            )

            time2 = time2.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model_siamese(
                    time1,
                    time2
                )

                loss = criterion(
                    outputs,
                    labels
                )


            val_losses.append(
                loss.item()
            )


            predictions = outputs.argmax(
                dim=1
            )


            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                predictions.cpu().numpy()
            )


    val_loss = np.mean(
        val_losses
    )

    val_acc = accuracy_score(
        val_true,
        val_pred
    )

    val_macro_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # Scheduler
    # --------------------------------------------------------

    scheduler.step(
        val_macro_f1
    )

    current_lr = optimizer.param_groups[0]["lr"]


    # --------------------------------------------------------
    # Save history
    # --------------------------------------------------------

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "train_macro_f1": train_macro_f1,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_macro_f1,
        "learning_rate": current_lr
    })


    # --------------------------------------------------------
    # Save BEST model
    # --------------------------------------------------------

    if val_macro_f1 > best_macro_f1:

        best_macro_f1 = val_macro_f1
        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict":
                    model_siamese.state_dict(),
                "val_accuracy":
                    val_acc,
                "val_macro_f1":
                    val_macro_f1
            },
            best_siamese_path
        )

        status = " ⭐ BEST SAVED"

    else:

        epochs_without_improvement += 1

        status = ""


    # --------------------------------------------------------
    # Print epoch result
    # --------------------------------------------------------

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Train F1: {train_macro_f1:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_macro_f1:.4f} | "
        f"LR: {current_lr:.6f}"
        f"{status}"
    )


    # --------------------------------------------------------
    # Early stopping
    # --------------------------------------------------------

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):

        print(
            "\nEarly stopping triggered."
        )

        break


# ============================================================
# 11. SAVE TRAINING HISTORY
# ============================================================

history_df = pd.DataFrame(
    history
)

history_df.to_csv(
    history_path,
    index=False
)


# ============================================================
# FINAL RESULT
# ============================================================

print("\n======================================")
print("SIAMESE MRI-ONLY TRAINING COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_macro_f1, 4)
)

print(
    "Best model:",
    best_siamese_path
)

print(
    "Training history:",
    history_path
)
```

```python
# ============================================================
# STEP 12.2: FINAL SIAMESE + CLINICAL FUSION MODEL
#
# Purpose:
#   Time1 MRI -> Shared Encoder -> F1
#   Time2 MRI -> Shared Encoder -> F2
#
#   Longitudinal change:
#       Delta_F = F2 - F1
#
#   MRI longitudinal features + Clinical features
#   (Age + Sex + Diagnosis) fuse করা।
#
#   একটি patient-specific latent representation তৈরি হবে,
#   যা আমাদের two-timepoint prototype Digital Twin state
#   হিসেবে ব্যবহার করা হবে।
#
# Output:
#   1. Progression logits -> 3 classes
#   2. 128-dimensional latent patient representation
# ============================================================

import os
import torch
import torch.nn as nn

from torchvision.models import (
    resnet18,
    ResNet18_Weights
)


class SiameseClinicalGliomaCNN(nn.Module):

    def __init__(
        self,
        clinical_dim=11,
        num_classes=3
    ):

        super().__init__()


        # ====================================================
        # 1. SHARED PRETRAINED MRI ENCODER
        # ====================================================

        backbone = resnet18(
            weights=ResNet18_Weights.DEFAULT
        )

        old_conv = backbone.conv1

        backbone.conv1 = nn.Conv2d(
            in_channels=2,
            out_channels=old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding,
            bias=False
        )


        # Initialize 2-channel MRI input layer
        with torch.no_grad():

            mean_weight = old_conv.weight.mean(
                dim=1,
                keepdim=True
            )

            backbone.conv1.weight.copy_(
                mean_weight.repeat(1, 2, 1, 1)
            )


        self.feature_dim = backbone.fc.in_features

        backbone.fc = nn.Identity()

        self.encoder = backbone


        # ====================================================
        # 2. CLINICAL FEATURE ENCODER
        # ====================================================

        self.clinical_encoder = nn.Sequential(

            nn.Linear(
                clinical_dim,
                32
            ),

            nn.ReLU(),

            nn.Dropout(0.2)
        )


        # ====================================================
        # 3. MRI LONGITUDINAL FEATURE SIZE
        #
        # F1 + F2 + (F2-F1)
        #
        # 512 + 512 + 512 = 1536
        # ====================================================

        mri_feature_dim = (
            self.feature_dim * 3
        )


        # MRI + Clinical
        fusion_input_dim = (
            mri_feature_dim + 32
        )


        # ====================================================
        # 4. MULTIMODAL FUSION
        #
        # Output = 128-dimensional patient-specific
        # latent representation
        # ====================================================

        self.fusion = nn.Sequential(

            nn.Linear(
                fusion_input_dim,
                256
            ),

            nn.LayerNorm(256),

            nn.ReLU(),

            nn.Dropout(0.4),

            nn.Linear(
                256,
                128
            ),

            nn.ReLU()
        )


        # ====================================================
        # 5. FINAL PROGRESSION CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Dropout(0.3),

            nn.Linear(
                128,
                num_classes
            )
        )


    # ========================================================
    # MRI PATIENT ENCODING
    # ========================================================

    def encode_patient(self, x):

        # x:
        # [B, 48, 2, 224, 224]

        B, S, C, H, W = x.shape

        x = x.reshape(
            B * S,
            C,
            H,
            W
        )


        # Slice features
        features = self.encoder(
            x
        )

        features = features.reshape(
            B,
            S,
            self.feature_dim
        )


        # 48 slices -> one patient MRI representation
        patient_feature = features.mean(
            dim=1
        )

        return patient_feature


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(
        self,
        time1,
        time2,
        clinical,
        return_latent=False
    ):

        # --------------------------------
        # Shared MRI encoder
        # --------------------------------

        F1 = self.encode_patient(
            time1
        )

        F2 = self.encode_patient(
            time2
        )


        # --------------------------------
        # Longitudinal change
        # --------------------------------

        delta_F = F2 - F1


        # --------------------------------
        # MRI longitudinal representation
        # --------------------------------

        mri_feature = torch.cat(
            [
                F1,
                F2,
                delta_F
            ],
            dim=1
        )


        # --------------------------------
        # Clinical representation
        # --------------------------------

        clinical_feature = (
            self.clinical_encoder(
                clinical
            )
        )


        # --------------------------------
        # MRI + Clinical fusion
        # --------------------------------

        combined = torch.cat(
            [
                mri_feature,
                clinical_feature
            ],
            dim=1
        )


        # --------------------------------
        # Patient-specific latent state
        # --------------------------------

        latent_state = self.fusion(
            combined
        )


        # --------------------------------
        # Prediction
        # --------------------------------

        logits = self.classifier(
            latent_state
        )


        if return_latent:
            return logits, latent_state

        return logits


# ============================================================
# 6. CREATE MULTIMODAL MODEL
# ============================================================

model_multimodal = SiameseClinicalGliomaCNN(
    clinical_dim=11,
    num_classes=3
).to(device)


# ============================================================
# 7. LOAD BEST MRI-ONLY SIAMESE ENCODER WEIGHTS
#
# আমরা MRI feature extraction আবার zero থেকে শিখছি না।
# Best Siamese MRI model-এর encoder reuse করছি।
# ============================================================

best_siamese_path = os.path.join(
    DATA_DIR,
    "research_models",
    "best_siamese_mri_only.pt"
)

checkpoint = torch.load(
    best_siamese_path,
    map_location=device
)

old_state = checkpoint[
    "model_state_dict"
]


# Only encoder weights
encoder_state = {
    key.replace("encoder.", ""): value
    for key, value in old_state.items()
    if key.startswith("encoder.")
}


model_multimodal.encoder.load_state_dict(
    encoder_state
)


print(
    "✅ Best Siamese MRI encoder weights loaded."
)


# ============================================================
# 8. SANITY CHECK
# ============================================================

model_multimodal.eval()

with torch.no_grad():

    t1, t2, clinical, _ = next(
        iter(train_multimodal_loader)
    )

    t1 = t1.to(device)
    t2 = t2.to(device)
    clinical = clinical.to(device)


    logits, latent = model_multimodal(
        t1,
        t2,
        clinical,
        return_latent=True
    )


print(
    "Prediction output:",
    logits.shape
)

print(
    "Digital Twin latent state:",
    latent.shape
)

print(
    "\n✅ Siamese + Clinical model ready."
)
```

```python
# ============================================================
# STEP 12.3: FINAL SIAMESE + CLINICAL MODEL TRAINING
#
# Purpose:
#   Time1 MRI + Time2 MRI + Clinical information ব্যবহার করে
#   final multimodal model train করা।
#
# Training strategy:
#   Epoch 1-3  : MRI encoder frozen
#   Epoch 4+   : Last ResNet block fine-tuned
#
# Best model selected by Validation Macro-F1.
# Test set এখানে ব্যবহার করা হচ্ছে না.
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import accuracy_score, f1_score


# ------------------------------------------------------------
# 1. Freeze MRI encoder initially
# ------------------------------------------------------------

for param in model_multimodal.encoder.parameters():
    param.requires_grad = False

for param in model_multimodal.clinical_encoder.parameters():
    param.requires_grad = True

for param in model_multimodal.fusion.parameters():
    param.requires_grad = True

for param in model_multimodal.classifier.parameters():
    param.requires_grad = True


# ------------------------------------------------------------
# 2. Weighted loss
# ------------------------------------------------------------

weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=weights_tensor,
    label_smoothing=0.05
)


# ------------------------------------------------------------
# 3. Optimizer
# ------------------------------------------------------------

optimizer = torch.optim.AdamW(
    [
        {
            "params": model_multimodal.clinical_encoder.parameters(),
            "lr": 1e-4
        },
        {
            "params": model_multimodal.fusion.parameters(),
            "lr": 1e-4
        },
        {
            "params": model_multimodal.classifier.parameters(),
            "lr": 1e-4
        }
    ],
    weight_decay=1e-4
)


# ------------------------------------------------------------
# 4. Mixed precision
# ------------------------------------------------------------

scaler = torch.amp.GradScaler("cuda")


# ------------------------------------------------------------
# 5. Paired augmentation
# Same transformation for Time1 and Time2
# ------------------------------------------------------------

def paired_augmentation(t1, t2):

    # Same horizontal flip
    if torch.rand(1).item() < 0.5:

        t1 = torch.flip(t1, dims=[-1])
        t2 = torch.flip(t2, dims=[-1])

    # Same small intensity scaling
    scale = torch.empty(
        1,
        device=t1.device
    ).uniform_(0.95, 1.05)

    t1 = t1 * scale
    t2 = t2 * scale

    return t1, t2


# ------------------------------------------------------------
# 6. Training settings
# ------------------------------------------------------------

EPOCHS = 15
UNFREEZE_EPOCH = 4
EARLY_STOPPING_PATIENCE = 5

best_macro_f1 = -1.0
epochs_without_improvement = 0

history = []


# ------------------------------------------------------------
# 7. Save paths
# ------------------------------------------------------------

MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

best_multimodal_path = os.path.join(
    MODEL_DIR,
    "best_siamese_clinical.pt"
)

history_path = os.path.join(
    MODEL_DIR,
    "siamese_clinical_history.csv"
)


# ============================================================
# 8. TRAINING LOOP
# ============================================================

for epoch in range(EPOCHS):


    # --------------------------------------------------------
    # At epoch 4, fine-tune only last ResNet block
    # --------------------------------------------------------

    if epoch + 1 == UNFREEZE_EPOCH:

        print("\n🔓 Fine-tuning ResNet layer4...\n")

        for param in model_multimodal.encoder.layer4.parameters():
            param.requires_grad = True

        optimizer.add_param_group({
            "params": model_multimodal.encoder.layer4.parameters(),
            "lr": 1e-5
        })


    # ========================================================
    # TRAIN
    # ========================================================

    model_multimodal.train()

    train_losses = []
    train_true = []
    train_pred = []


    for t1, t2, clinical, labels in train_multimodal_loader:

        t1 = t1.to(device, non_blocking=True)
        t2 = t2.to(device, non_blocking=True)

        clinical = clinical.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        t1, t2 = paired_augmentation(
            t1,
            t2
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_multimodal(
                t1,
                t2,
                clinical
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(loss).backward()

        scaler.step(optimizer)

        scaler.update()


        train_losses.append(
            loss.item()
        )


        preds = outputs.argmax(
            dim=1
        )


        train_true.extend(
            labels.detach().cpu().numpy()
        )

        train_pred.extend(
            preds.detach().cpu().numpy()
        )


    train_loss = np.mean(
        train_losses
    )

    train_acc = accuracy_score(
        train_true,
        train_pred
    )

    train_f1 = f1_score(
        train_true,
        train_pred,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    model_multimodal.eval()

    val_losses = []
    val_true = []
    val_pred = []


    with torch.no_grad():

        for t1, t2, clinical, labels in val_multimodal_loader:

            t1 = t1.to(device)
            t2 = t2.to(device)
            clinical = clinical.to(device)
            labels = labels.to(device)


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model_multimodal(
                    t1,
                    t2,
                    clinical
                )

                loss = criterion(
                    outputs,
                    labels
                )


            val_losses.append(
                loss.item()
            )

            preds = outputs.argmax(
                dim=1
            )


            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                preds.cpu().numpy()
            )


    val_loss = np.mean(
        val_losses
    )

    val_acc = accuracy_score(
        val_true,
        val_pred
    )

    val_macro_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # Save history
    # --------------------------------------------------------

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "train_macro_f1": train_f1,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_macro_f1
    })


    # --------------------------------------------------------
    # Save best model
    # --------------------------------------------------------

    if val_macro_f1 > best_macro_f1:

        best_macro_f1 = val_macro_f1

        epochs_without_improvement = 0


        torch.save(
            {
                "epoch": epoch + 1,

                "model_state_dict":
                    model_multimodal.state_dict(),

                "val_accuracy":
                    val_acc,

                "val_macro_f1":
                    val_macro_f1
            },

            best_multimodal_path
        )

        status = " ⭐ BEST SAVED"


    else:

        epochs_without_improvement += 1

        status = ""


    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Train F1: {train_f1:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_macro_f1:.4f}"
        f"{status}"
    )


    # --------------------------------------------------------
    # Early stopping
    # --------------------------------------------------------

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):

        print("\nEarly stopping triggered.")
        break


# ============================================================
# 9. SAVE HISTORY
# ============================================================

history_df = pd.DataFrame(history)

history_df.to_csv(
    history_path,
    index=False
)


print("\n======================================")
print("SIAMESE + CLINICAL TRAINING COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_macro_f1, 4)
)

print(
    "Best model:",
    best_multimodal_path
)

print(
    "History:",
    history_path
)
```

```python
# ============================================================
# STEP 12.4: IMPROVED SIAMESE + CLINICAL + ATTENTION TRAINING
#
# Improvements:
#   1. Mean pooling -> Learnable Slice Attention Pooling
#   2. Same attention module for Time1 and Time2
#   3. F1, F2, F2-F1 preserved
#   4. MRI and Clinical features separately projected
#      before fusion so MRI does not dominate clinical data
#   5. Best MRI-only encoder weights reused
#   6. Weighted loss + early stopping
#
# Test set is NOT used.
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torchvision.models import resnet18, ResNet18_Weights
from sklearn.metrics import accuracy_score, f1_score


# ============================================================
# 1. ATTENTION-BASED MULTIMODAL MODEL
# ============================================================

class ImprovedSiameseClinicalCNN(nn.Module):

    def __init__(self, clinical_dim=11, num_classes=3):

        super().__init__()

        # ----------------------------------------------------
        # Shared ResNet18 encoder
        # ----------------------------------------------------

        backbone = resnet18(
            weights=ResNet18_Weights.DEFAULT
        )

        old_conv = backbone.conv1

        backbone.conv1 = nn.Conv2d(
            2,
            old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding,
            bias=False
        )

        with torch.no_grad():

            mean_weight = old_conv.weight.mean(
                dim=1,
                keepdim=True
            )

            backbone.conv1.weight.copy_(
                mean_weight.repeat(1, 2, 1, 1)
            )

        self.feature_dim = backbone.fc.in_features  # 512

        backbone.fc = nn.Identity()

        self.encoder = backbone


        # ----------------------------------------------------
        # Shared slice-attention module
        #
        # Learns which of the 48 MRI slices are important.
        # Same module used for Time1 and Time2.
        # ----------------------------------------------------

        self.slice_attention = nn.Sequential(

            nn.Linear(
                self.feature_dim,
                128
            ),

            nn.Tanh(),

            nn.Linear(
                128,
                1
            )
        )


        # ----------------------------------------------------
        # MRI projection
        #
        # F1 + F2 + DeltaF = 1536
        # 1536 -> 128
        # ----------------------------------------------------

        self.mri_projection = nn.Sequential(

            nn.Linear(
                self.feature_dim * 3,
                256
            ),

            nn.LayerNorm(256),

            nn.ReLU(),

            nn.Dropout(0.4),

            nn.Linear(
                256,
                128
            ),

            nn.ReLU()
        )


        # ----------------------------------------------------
        # Clinical projection
        #
        # Age + Sex + Diagnosis
        # 11 -> 64
        # ----------------------------------------------------

        self.clinical_projection = nn.Sequential(

            nn.Linear(
                clinical_dim,
                64
            ),

            nn.LayerNorm(64),

            nn.ReLU(),

            nn.Dropout(0.2)
        )


        # ----------------------------------------------------
        # Balanced MRI + Clinical fusion
        #
        # MRI = 128
        # Clinical = 64
        # Combined = 192
        #
        # Final latent state = 128
        # ----------------------------------------------------

        self.fusion = nn.Sequential(

            nn.Linear(
                128 + 64,
                128
            ),

            nn.LayerNorm(128),

            nn.ReLU(),

            nn.Dropout(0.4)
        )


        self.classifier = nn.Linear(
            128,
            num_classes
        )


    # ========================================================
    # Encode one timepoint
    # ========================================================

    def encode_patient(self, x):

        # x = [B, 48, 2, 224, 224]

        B, S, C, H, W = x.shape

        x = x.reshape(
            B * S,
            C,
            H,
            W
        )

        # Slice-level ResNet features
        features = self.encoder(x)

        features = features.reshape(
            B,
            S,
            self.feature_dim
        )


        # ----------------------------------------------------
        # Learn attention score for each slice
        # ----------------------------------------------------

        scores = self.slice_attention(
            features
        ).squeeze(-1)

        attention = torch.softmax(
            scores,
            dim=1
        )


        # ----------------------------------------------------
        # Weighted patient representation
        # ----------------------------------------------------

        patient_feature = torch.sum(
            features * attention.unsqueeze(-1),
            dim=1
        )


        return patient_feature, attention


    # ========================================================
    # Forward
    # ========================================================

    def forward(
        self,
        time1,
        time2,
        clinical,
        return_latent=False,
        return_attention=False
    ):

        # Shared encoder + shared attention
        F1, att1 = self.encode_patient(time1)
        F2, att2 = self.encode_patient(time2)


        # Proposal-aligned signed feature change
        delta_F = F2 - F1


        longitudinal_feature = torch.cat(
            [
                F1,
                F2,
                delta_F
            ],
            dim=1
        )


        # MRI representation
        mri_feature = self.mri_projection(
            longitudinal_feature
        )


        # Clinical representation
        clinical_feature = self.clinical_projection(
            clinical
        )


        # Multimodal fusion
        combined = torch.cat(
            [
                mri_feature,
                clinical_feature
            ],
            dim=1
        )


        latent_state = self.fusion(
            combined
        )


        logits = self.classifier(
            latent_state
        )


        if return_latent and return_attention:

            return (
                logits,
                latent_state,
                att1,
                att2
            )

        if return_latent:

            return logits, latent_state

        return logits


# ============================================================
# 2. CREATE MODEL
# ============================================================

model_improved = ImprovedSiameseClinicalCNN(
    clinical_dim=11,
    num_classes=3
).to(device)


# ============================================================
# 3. LOAD BEST MRI-ONLY ENCODER
# ============================================================

best_siamese_path = os.path.join(
    DATA_DIR,
    "research_models",
    "best_siamese_mri_only.pt"
)

checkpoint = torch.load(
    best_siamese_path,
    map_location=device
)

old_state = checkpoint["model_state_dict"]


encoder_state = {
    key.replace("encoder.", ""): value
    for key, value in old_state.items()
    if key.startswith("encoder.")
}


model_improved.encoder.load_state_dict(
    encoder_state
)

print("✅ Best MRI-only encoder loaded.")


# ============================================================
# 4. TRAINING STRATEGY
#
# First train attention + fusion.
# Fine-tune ResNet layer4 after epoch 3.
# ============================================================

for param in model_improved.encoder.parameters():
    param.requires_grad = False


# ------------------------------------------------------------
# Weighted loss
# ------------------------------------------------------------

weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=weights_tensor,
    label_smoothing=0.05
)


# ------------------------------------------------------------
# Optimizer
# ------------------------------------------------------------

head_parameters = (
    list(model_improved.slice_attention.parameters()) +
    list(model_improved.mri_projection.parameters()) +
    list(model_improved.clinical_projection.parameters()) +
    list(model_improved.fusion.parameters()) +
    list(model_improved.classifier.parameters())
)

optimizer = torch.optim.AdamW(
    head_parameters,
    lr=1e-4,
    weight_decay=2e-4
)


scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)


scaler = torch.amp.GradScaler("cuda")


# ============================================================
# 5. TRAINING SETTINGS
# ============================================================

EPOCHS = 20
UNFREEZE_EPOCH = 4
PATIENCE = 6

best_f1 = -1.0
no_improvement = 0

history = []


MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

best_path = os.path.join(
    MODEL_DIR,
    "best_improved_siamese_clinical.pt"
)

history_path = os.path.join(
    MODEL_DIR,
    "improved_siamese_clinical_history.csv"
)


# ============================================================
# 6. TRAIN
# ============================================================

for epoch in range(EPOCHS):


    # --------------------------------------------------------
    # Controlled fine-tuning after epoch 3
    # --------------------------------------------------------

    if epoch + 1 == UNFREEZE_EPOCH:

        print("\n🔓 Fine-tuning ResNet layer4...\n")

        for param in model_improved.encoder.layer4.parameters():
            param.requires_grad = True


        optimizer.add_param_group({
            "params":
                model_improved.encoder.layer4.parameters(),
            "lr":
                1e-5
        })


    # ========================================================
    # TRAIN PHASE
    # ========================================================

    model_improved.train()

    train_losses = []
    train_true = []
    train_pred = []


    for t1, t2, clinical, labels in train_multimodal_loader:

        t1 = t1.to(device, non_blocking=True)
        t2 = t2.to(device, non_blocking=True)

        clinical = clinical.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        # Same geometric augmentation
        if torch.rand(1).item() < 0.5:

            t1 = torch.flip(
                t1,
                dims=[-1]
            )

            t2 = torch.flip(
                t2,
                dims=[-1]
            )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_improved(
                t1,
                t2,
                clinical
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(loss).backward()

        scaler.step(optimizer)

        scaler.update()


        train_losses.append(
            loss.item()
        )


        preds = outputs.argmax(
            dim=1
        )


        train_true.extend(
            labels.detach().cpu().numpy()
        )

        train_pred.extend(
            preds.detach().cpu().numpy()
        )


    train_loss = np.mean(
        train_losses
    )

    train_acc = accuracy_score(
        train_true,
        train_pred
    )

    train_f1 = f1_score(
        train_true,
        train_pred,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # VALIDATION PHASE
    # ========================================================

    model_improved.eval()

    val_losses = []
    val_true = []
    val_pred = []


    with torch.no_grad():

        for t1, t2, clinical, labels in val_multimodal_loader:

            t1 = t1.to(device)
            t2 = t2.to(device)
            clinical = clinical.to(device)
            labels = labels.to(device)


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model_improved(
                    t1,
                    t2,
                    clinical
                )

                loss = criterion(
                    outputs,
                    labels
                )


            val_losses.append(
                loss.item()
            )


            preds = outputs.argmax(
                dim=1
            )


            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                preds.cpu().numpy()
            )


    val_loss = np.mean(
        val_losses
    )

    val_acc = accuracy_score(
        val_true,
        val_pred
    )

    val_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    scheduler.step(
        val_f1
    )


    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "train_macro_f1": train_f1,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_f1
    })


    # --------------------------------------------------------
    # Save best model
    # --------------------------------------------------------

    if val_f1 > best_f1:

        best_f1 = val_f1
        no_improvement = 0


        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict":
                    model_improved.state_dict(),
                "val_accuracy":
                    val_acc,
                "val_macro_f1":
                    val_f1
            },
            best_path
        )

        status = " ⭐ BEST SAVED"


    else:

        no_improvement += 1
        status = ""


    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Train F1: {train_f1:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_f1:.4f}"
        f"{status}"
    )


    if no_improvement >= PATIENCE:

        print("\nEarly stopping triggered.")
        break


# ============================================================
# 7. SAVE HISTORY
# ============================================================

pd.DataFrame(
    history
).to_csv(
    history_path,
    index=False
)


print("\n======================================")
print("IMPROVED MULTIMODAL TRAINING COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_f1, 4)
)

print(
    "Best model:",
    best_path
)

print(
    "History:",
    history_path
)
```

```python
# ============================================================
# STEP 12.6: BALANCED TRAINING FOR RARE "DECREASED" CLASS
#
# Why:
#   Previous model predicted ZERO Decreased patients.
#
# Strategy:
#   1. WeightedRandomSampler -> balanced training
#   2. Encoder stays FROZEN
#   3. Attention + fusion head only trained
#   4. Best model selected by Validation Macro-F1
#
# Test set is NOT used.
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import accuracy_score, f1_score


# ============================================================
# 1. BALANCED SAMPLER
# ============================================================

train_labels = [
    label_map[label]
    for label in train_df["Label"]
]

class_counts = np.bincount(
    train_labels
)

print("Train class counts:", class_counts)


# Inverse-frequency weight
class_sample_weights = 1.0 / class_counts

sample_weights = [
    class_sample_weights[label]
    for label in train_labels
]

sampler = WeightedRandomSampler(
    weights=torch.DoubleTensor(sample_weights),
    num_samples=len(sample_weights),
    replacement=True
)


train_balanced_loader = DataLoader(
    train_multimodal_dataset,
    batch_size=1,
    sampler=sampler,
    num_workers=2,
    pin_memory=True
)

print("✅ Balanced DataLoader ready.")


# ============================================================
# 2. NEW ATTENTION MODEL
# ============================================================

model_balanced = ImprovedSiameseClinicalCNN(
    clinical_dim=11,
    num_classes=3
).to(device)


# ============================================================
# 3. LOAD BEST MRI-ONLY ENCODER
# ============================================================

best_siamese_path = os.path.join(
    DATA_DIR,
    "research_models",
    "best_siamese_mri_only.pt"
)

checkpoint = torch.load(
    best_siamese_path,
    map_location=device
)

old_state = checkpoint[
    "model_state_dict"
]

encoder_state = {
    key.replace("encoder.", ""): value
    for key, value in old_state.items()
    if key.startswith("encoder.")
}

model_balanced.encoder.load_state_dict(
    encoder_state
)

print("✅ MRI encoder loaded.")


# ============================================================
# 4. KEEP ENCODER FROZEN
# ============================================================

for param in model_balanced.encoder.parameters():
    param.requires_grad = False


# ============================================================
# 5. LOSS
#
# Balanced sampler already handles imbalance,
# so we DO NOT use strong class weights again.
# ============================================================

criterion = nn.CrossEntropyLoss(
    label_smoothing=0.05
)


# ============================================================
# 6. OPTIMIZER
# ============================================================

trainable_params = [
    p for p in model_balanced.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=1e-4,
    weight_decay=2e-4
)


scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)

scaler = torch.amp.GradScaler("cuda")


# ============================================================
# 7. SETTINGS
# ============================================================

EPOCHS = 20
PATIENCE = 6

best_f1 = -1.0
no_improve = 0

history = []

best_path = os.path.join(
    DATA_DIR,
    "research_models",
    "best_balanced_siamese_clinical.pt"
)


# ============================================================
# 8. TRAINING
# ============================================================

for epoch in range(EPOCHS):

    # ---------------- TRAIN ----------------

    model_balanced.train()

    train_true = []
    train_pred = []
    train_losses = []


    for t1, t2, clinical, labels in train_balanced_loader:

        t1 = t1.to(device)
        t2 = t2.to(device)
        clinical = clinical.to(device)
        labels = labels.to(device)


        # Same paired flip
        if torch.rand(1).item() < 0.5:

            t1 = torch.flip(t1, dims=[-1])
            t2 = torch.flip(t2, dims=[-1])


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_balanced(
                t1,
                t2,
                clinical
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()


        train_losses.append(
            loss.item()
        )

        preds = outputs.argmax(dim=1)

        train_true.extend(
            labels.cpu().numpy()
        )

        train_pred.extend(
            preds.detach().cpu().numpy()
        )


    train_acc = accuracy_score(
        train_true,
        train_pred
    )

    train_f1 = f1_score(
        train_true,
        train_pred,
        average="macro",
        zero_division=0
    )


    # ---------------- VALIDATION ----------------

    model_balanced.eval()

    val_true = []
    val_pred = []


    with torch.no_grad():

        for t1, t2, clinical, labels in val_multimodal_loader:

            t1 = t1.to(device)
            t2 = t2.to(device)
            clinical = clinical.to(device)


            outputs = model_balanced(
                t1,
                t2,
                clinical
            )

            preds = outputs.argmax(
                dim=1
            )


            val_true.extend(
                labels.numpy()
            )

            val_pred.extend(
                preds.cpu().numpy()
            )


    val_acc = accuracy_score(
        val_true,
        val_pred
    )

    val_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    scheduler.step(
        val_f1
    )


    # ---------------- SAVE BEST ----------------

    if val_f1 > best_f1:

        best_f1 = val_f1
        no_improve = 0

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict":
                    model_balanced.state_dict(),
                "val_accuracy":
                    val_acc,
                "val_macro_f1":
                    val_f1
            },
            best_path
        )

        status = " ⭐ BEST SAVED"

    else:

        no_improve += 1
        status = ""


    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Train F1: {train_f1:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_f1:.4f}"
        f"{status}"
    )


    if no_improve >= PATIENCE:

        print("\nEarly stopping triggered.")
        break


print("\n======================================")
print("BALANCED TRAINING COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_f1, 4)
)

print(
    "Best model:",
    best_path
)
```

```python
# ============================================================
# STEP 16-17: TIME1 + TIME2 GRAD-CAM
#
# Purpose:
#   Final Siamese + Clinical model-এর prediction করার সময়
#   Time1 এবং Time2 MRI-তে model কোথায় attention দিয়েছে
#   তা Grad-CAM দিয়ে দেখা।
#
# Grad-CAM target:
#   Shared ResNet18 -> layer4[-1]
#
# Output:
#   - Time1 Grad-CAM
#   - Time2 Grad-CAM
#   - MRI + Grad-CAM overlay
#   - Example figure saved to Google Drive
# ============================================================

import os
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Class names
# ------------------------------------------------------------

class_names = [
    "Unchanged",
    "Increased",
    "Decreased"
]


# ------------------------------------------------------------
# 2. Grad-CAM function
# ------------------------------------------------------------

def generate_longitudinal_gradcam(
    model,
    time1,
    time2,
    clinical
):

    model.eval()
    model.zero_grad(set_to_none=True)

    # Gradients যেন frozen encoder হলেও পাওয়া যায়
    time1 = time1.clone().detach().requires_grad_(True)
    time2 = time2.clone().detach().requires_grad_(True)

    captures = []


    # --------------------------------------------------------
    # Hook shared encoder target layer
    # Encoder দুইবার call হবে:
    # captures[0] = Time1
    # captures[1] = Time2
    # --------------------------------------------------------

    def forward_hook(module, inputs, output):

        record = {
            "activation": output.detach()
        }

        def save_gradient(grad):

            record["gradient"] = grad.detach()

        output.register_hook(
            save_gradient
        )

        captures.append(
            record
        )


    handle = model.encoder.layer4[-1].register_forward_hook(
        forward_hook
    )


    # --------------------------------------------------------
    # Forward
    # --------------------------------------------------------

    logits = model(
        time1,
        time2,
        clinical
    )

    predicted_class = logits.argmax(
        dim=1
    ).item()


    probability = torch.softmax(
        logits,
        dim=1
    )[0, predicted_class].item()


    # --------------------------------------------------------
    # Backward predicted class
    # --------------------------------------------------------

    score = logits[
        0,
        predicted_class
    ]

    score.backward()

    handle.remove()


    # --------------------------------------------------------
    # Grad-CAM calculation
    # --------------------------------------------------------

    def make_cam(record):

        activation = record[
            "activation"
        ]

        gradient = record[
            "gradient"
        ]


        # [48, 512, 7, 7]
        weights = gradient.mean(
            dim=(2, 3),
            keepdim=True
        )


        cam = (
            weights * activation
        ).sum(
            dim=1
        )


        cam = torch.relu(
            cam
        )


        # 7x7 -> 224x224
        cam = F.interpolate(
            cam.unsqueeze(1),
            size=(224, 224),
            mode="bilinear",
            align_corners=False
        ).squeeze(1)


        # Normalize entire patient CAM to 0-1
        cam_min = cam.min()
        cam_max = cam.max()

        cam = (
            cam - cam_min
        ) / (
            cam_max - cam_min + 1e-8
        )


        return cam.detach().cpu().numpy()


    cam_time1 = make_cam(
        captures[0]
    )

    cam_time2 = make_cam(
        captures[1]
    )


    return (
        cam_time1,
        cam_time2,
        predicted_class,
        probability
    )


# ============================================================
# 3. SELECT ONE TEST PATIENT
#
# Deterministic example: first test patient
# Quantitative analysis পরে ALL patients-এর ওপর হবে.
# ============================================================

example_index = 0

t1, t2, clinical, true_label = (
    test_multimodal_dataset[
        example_index
    ]
)

patient_id = str(
    test_df.iloc[
        example_index
    ]["PatientID"]
)


# Add batch dimension
t1_batch = t1.unsqueeze(0).to(device)
t2_batch = t2.unsqueeze(0).to(device)

clinical_batch = clinical.unsqueeze(0).to(
    device
)


# ============================================================
# 4. GENERATE GRAD-CAM
# ============================================================

cam1, cam2, pred_class, probability = (
    generate_longitudinal_gradcam(
        model_improved,
        t1_batch,
        t2_batch,
        clinical_batch
    )
)


print(
    "Patient ID:",
    patient_id
)

print(
    "True class:",
    class_names[
        int(true_label)
    ]
)

print(
    "Predicted class:",
    class_names[
        pred_class
    ]
)

print(
    "Prediction probability:",
    round(
        probability,
        4
    )
)


# ============================================================
# 5. SELECT MOST ATTENDED SLICE
#
# Mask ব্যবহার করে slice select করছি না।
# Grad-CAM attention দিয়েই slice select হচ্ছে.
# ============================================================

slice_scores = (
    cam1.mean(axis=(1, 2)) +
    cam2.mean(axis=(1, 2))
)

best_slice = int(
    np.argmax(
        slice_scores
    )
)


print(
    "Most attended processed slice:",
    best_slice
)


# ============================================================
# 6. MRI IMAGES
#
# Channel:
#   0 = FLAIR
#   1 = T1CE
# ============================================================

img1 = t1[
    best_slice,
    1
].numpy()

img2 = t2[
    best_slice,
    1
].numpy()


# ============================================================
# 7. VISUALIZE
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(10, 9)
)


# Time1 MRI
axes[0, 0].imshow(
    img1,
    cmap="gray"
)

axes[0, 0].set_title(
    "Time1 T1CE"
)

axes[0, 0].axis("off")


# Time1 overlay
axes[0, 1].imshow(
    img1,
    cmap="gray"
)

axes[0, 1].imshow(
    cam1[best_slice],
    cmap="jet",
    alpha=0.45
)

axes[0, 1].set_title(
    "Time1 Grad-CAM"
)

axes[0, 1].axis("off")


# Time2 MRI
axes[1, 0].imshow(
    img2,
    cmap="gray"
)

axes[1, 0].set_title(
    "Time2 T1CE"
)

axes[1, 0].axis("off")


# Time2 overlay
axes[1, 1].imshow(
    img2,
    cmap="gray"
)

axes[1, 1].imshow(
    cam2[best_slice],
    cmap="jet",
    alpha=0.45
)

axes[1, 1].set_title(
    "Time2 Grad-CAM"
)

axes[1, 1].axis("off")


plt.suptitle(
    f"Patient {patient_id} | "
    f"True: {class_names[int(true_label)]} | "
    f"Predicted: {class_names[pred_class]}"
)

plt.tight_layout()


# ============================================================
# 8. SAVE FIGURE
# ============================================================

OUTPUT_DIR = os.path.join(
    DATA_DIR,
    "research_outputs"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

gradcam_path = os.path.join(
    OUTPUT_DIR,
    f"gradcam_example_{patient_id}.png"
)

plt.savefig(
    gradcam_path,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


print(
    "\n✅ Grad-CAM saved at:"
)

print(
    gradcam_path
)
```

```python
# ============================================================
# STEP 17.1: CLEAN LONGITUDINAL GRAD-CAM
#
# Fixes:
#   1. Remove Grad-CAM outside brain
#   2. Joint normalization of Time1 + Time2
#      -> important for temporal comparison
#   3. Select representative slice using brain-only attention
#
# Output:
#   clean_cam1
#   clean_cam2
# ============================================================

import numpy as np
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Create brain masks from processed MRI
# Background is zero in our preprocessed images
# ------------------------------------------------------------

t1_np = t1.numpy()      # [48, 2, 224, 224]
t2_np = t2.numpy()


brain1 = np.any(
    np.abs(t1_np) > 1e-6,
    axis=1
)

brain2 = np.any(
    np.abs(t2_np) > 1e-6,
    axis=1
)


# Same spatial support for longitudinal comparison
brain_union = brain1 | brain2


# ------------------------------------------------------------
# 2. Remove background CAM
# ------------------------------------------------------------

clean_cam1 = cam1 * brain_union
clean_cam2 = cam2 * brain_union


# ------------------------------------------------------------
# 3. JOINT normalization
#
# Important:
# We do NOT separately rescale T1 and T2.
# One common maximum keeps temporal magnitude comparable.
# ------------------------------------------------------------

joint_max = max(
    clean_cam1.max(),
    clean_cam2.max()
)

if joint_max > 0:

    clean_cam1 = clean_cam1 / joint_max
    clean_cam2 = clean_cam2 / joint_max


# ------------------------------------------------------------
# 4. Select strongest brain-only attention slice
# ------------------------------------------------------------

joint_slice_score = (
    clean_cam1.mean(axis=(1, 2)) +
    clean_cam2.mean(axis=(1, 2))
)

best_clean_slice = int(
    np.argmax(joint_slice_score)
)


print(
    "Clean Grad-CAM slice:",
    best_clean_slice
)

print(
    "Time1 CAM max:",
    round(float(clean_cam1.max()), 4)
)

print(
    "Time2 CAM max:",
    round(float(clean_cam2.max()), 4)
)


# ------------------------------------------------------------
# 5. Visualize corrected CAM
# ------------------------------------------------------------

img1 = t1_np[
    best_clean_slice,
    1
]

img2 = t2_np[
    best_clean_slice,
    1
]


fig, axes = plt.subplots(
    1,
    2,
    figsize=(10, 5)
)


axes[0].imshow(
    img1,
    cmap="gray"
)

axes[0].imshow(
    clean_cam1[best_clean_slice],
    cmap="jet",
    alpha=0.45,
    vmin=0,
    vmax=1
)

axes[0].set_title(
    "Clean Time1 Grad-CAM"
)

axes[0].axis("off")


axes[1].imshow(
    img2,
    cmap="gray"
)

axes[1].imshow(
    clean_cam2[best_clean_slice],
    cmap="jet",
    alpha=0.45,
    vmin=0,
    vmax=1
)

axes[1].set_title(
    "Clean Time2 Grad-CAM"
)

axes[1].axis("off")


plt.suptitle(
    f"Patient {patient_id} | "
    f"{class_names[pred_class]}"
)

plt.tight_layout()
plt.show()


print(
    "\n✅ Clean longitudinal Grad-CAM ready."
)
```

```python
# ============================================================
# STEP 17.3: SIGNED LONGITUDINAL GRAD-CAM
#
# Why:
#   Standard Grad-CAM uses ReLU and removed negative evidence.
#   In our Siamese model Delta_F = F2 - F1, so negative
#   branch contributions are meaningful.
#
# Quantitative XAI will therefore use:
#
#   Signed CAM     -> direction of contribution
#   |Signed CAM|   -> spatial attention magnitude
#
# IMPORTANT:
#   No tumor masks are used to create these maps.
#   Therefore there is no segmentation-mask leakage.
# ============================================================

import numpy as np
import torch
import torch.nn.functional as F


def generate_signed_gradcam(
    model,
    time1,
    time2,
    clinical
):

    model.eval()
    model.zero_grad(set_to_none=True)

    time1 = (
        time1.clone()
        .detach()
        .requires_grad_(True)
    )

    time2 = (
        time2.clone()
        .detach()
        .requires_grad_(True)
    )

    captures = []


    # --------------------------------------------------------
    # Capture Time1 and Time2 activations + gradients
    # --------------------------------------------------------

    def hook_function(module, inputs, output):

        record = {
            "activation": output.detach()
        }

        def save_gradient(grad):

            record["gradient"] = grad.detach()

        output.register_hook(
            save_gradient
        )

        captures.append(
            record
        )


    handle = (
        model.encoder.layer4[-1]
        .register_forward_hook(
            hook_function
        )
    )


    # --------------------------------------------------------
    # Forward prediction
    # --------------------------------------------------------

    logits = model(
        time1,
        time2,
        clinical
    )

    predicted_class = int(
        logits.argmax(dim=1).item()
    )

    probability = float(
        torch.softmax(
            logits,
            dim=1
        )[0, predicted_class].item()
    )


    # --------------------------------------------------------
    # Backpropagate predicted-class score
    # --------------------------------------------------------

    score = logits[
        0,
        predicted_class
    ]

    score.backward()

    handle.remove()


    # --------------------------------------------------------
    # Raw SIGNED Grad-CAM
    #
    # Notice:
    # NO torch.relu() here.
    # --------------------------------------------------------

    def build_signed_cam(record):

        activation = record[
            "activation"
        ]

        gradient = record[
            "gradient"
        ]


        weights = gradient.mean(
            dim=(2, 3),
            keepdim=True
        )


        raw_cam = (
            weights * activation
        ).sum(
            dim=1
        )


        # 7x7 -> 224x224
        raw_cam = F.interpolate(
            raw_cam.unsqueeze(1),
            size=(224, 224),
            mode="bilinear",
            align_corners=False
        ).squeeze(1)


        return (
            raw_cam
            .detach()
            .cpu()
            .numpy()
        )


    raw1 = build_signed_cam(
        captures[0]
    )

    raw2 = build_signed_cam(
        captures[1]
    )


    # --------------------------------------------------------
    # Brain mask
    # --------------------------------------------------------

    t1_np = (
        time1[0]
        .detach()
        .cpu()
        .numpy()
    )

    t2_np = (
        time2[0]
        .detach()
        .cpu()
        .numpy()
    )


    brain1 = np.any(
        np.abs(t1_np) > 1e-6,
        axis=1
    )

    brain2 = np.any(
        np.abs(t2_np) > 1e-6,
        axis=1
    )

    brain_union = (
        brain1 | brain2
    )


    raw1 = raw1 * brain_union
    raw2 = raw2 * brain_union


    # --------------------------------------------------------
    # Save raw magnitude BEFORE normalization for QC
    # --------------------------------------------------------

    raw_max1 = float(
        np.max(
            np.abs(raw1)
        )
    )

    raw_max2 = float(
        np.max(
            np.abs(raw2)
        )
    )


    # --------------------------------------------------------
    # JOINT normalization
    #
    # One scale for BOTH timepoints.
    # Signed output range approximately [-1, +1].
    # --------------------------------------------------------

    joint_abs_max = max(
        raw_max1,
        raw_max2,
        1e-12
    )


    signed1 = (
        raw1 /
        joint_abs_max
    )

    signed2 = (
        raw2 /
        joint_abs_max
    )


    # Spatial attention magnitude
    magnitude1 = np.abs(
        signed1
    )

    magnitude2 = np.abs(
        signed2
    )


    return {
        "signed_t1": signed1,
        "signed_t2": signed2,

        "magnitude_t1": magnitude1,
        "magnitude_t2": magnitude2,

        "raw_max_t1": raw_max1,
        "raw_max_t2": raw_max2,

        "prediction": predicted_class,
        "probability": probability
    }


# ============================================================
# QC ON ALL 45 TEST PATIENTS
# ============================================================

signed_zero_t1 = []
signed_zero_t2 = []

qc_records = []


for i in range(
    len(test_multimodal_dataset)
):

    t1_i, t2_i, clinical_i, label_i = (
        test_multimodal_dataset[i]
    )

    pid = str(
        test_df.iloc[i][
            "PatientID"
        ]
    )


    result = generate_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(
            device
        ),

        t2_i.unsqueeze(0).to(
            device
        ),

        clinical_i.unsqueeze(0).to(
            device
        )
    )


    max1 = result[
        "raw_max_t1"
    ]

    max2 = result[
        "raw_max_t2"
    ]


    if max1 <= 1e-12:

        signed_zero_t1.append(
            pid
        )


    if max2 <= 1e-12:

        signed_zero_t2.append(
            pid
        )


    qc_records.append(
        [
            pid,
            max1,
            max2
        ]
    )


print(
    "======================================"
)

print(
    "SIGNED GRAD-CAM QC"
)

print(
    "======================================"
)

print(
    "Total test patients:",
    len(test_multimodal_dataset)
)

print(
    "Zero Time1 Signed CAM:",
    len(signed_zero_t1)
)

print(
    "Zero Time2 Signed CAM:",
    len(signed_zero_t2)
)

print(
    "\nTime1 zero patients:",
    signed_zero_t1
)

print(
    "\nTime2 zero patients:",
    signed_zero_t2
)
```

```python
# ============================================================
# STEP 18: GRAD-CAM vs EXPERT ET MASK
#
# Purpose:
#   Signed Grad-CAM magnitude এবং expert ET segmentation mask
#   quantitatively compare করা।
#
# Metrics:
#   - Dice
#   - IoU
#   - Attention Precision
#   - Attention Recall
#
# ET label = 3
#
# Attention threshold:
#   Top 20% attention inside brain
#
# IMPORTANT:
#   Threshold test performance দেখে tune করা হয়নি.
# ============================================================

import os
import numpy as np
import pandas as pd

from tqdm.auto import tqdm


# ------------------------------------------------------------
# 1. Metric function
# ------------------------------------------------------------

def calculate_overlap_metrics(attention_mask, tumor_mask):

    attention_mask = attention_mask.astype(bool)
    tumor_mask = tumor_mask.astype(bool)

    intersection = np.logical_and(
        attention_mask,
        tumor_mask
    ).sum()

    attention_size = attention_mask.sum()
    tumor_size = tumor_mask.sum()

    union = np.logical_or(
        attention_mask,
        tumor_mask
    ).sum()


    # Dice
    if attention_size + tumor_size == 0:
        dice = 1.0
    else:
        dice = (
            2.0 * intersection
        ) / (
            attention_size + tumor_size
        )


    # IoU
    if union == 0:
        iou = 1.0
    else:
        iou = intersection / union


    # Attention precision:
    # কত attention tumor-এর ভিতরে পড়েছে?
    if attention_size == 0:
        precision = 0.0
    else:
        precision = (
            intersection /
            attention_size
        )


    # Attention recall:
    # আসল tumor-এর কত অংশ attention cover করেছে?
    if tumor_size == 0:
        recall = np.nan
    else:
        recall = (
            intersection /
            tumor_size
        )


    return (
        dice,
        iou,
        precision,
        recall
    )


# ------------------------------------------------------------
# 2. Convert continuous CAM -> binary attention mask
#
# Fixed rule:
# Top 20% attention inside brain
# ------------------------------------------------------------

def make_attention_mask(
    magnitude_cam,
    brain_mask,
    percentile=80
):

    values = magnitude_cam[
        brain_mask
    ]

    # Remove exact zeros
    positive_values = values[
        values > 0
    ]


    if len(positive_values) == 0:

        return np.zeros_like(
            magnitude_cam,
            dtype=bool
        ), np.nan


    threshold = np.percentile(
        positive_values,
        percentile
    )


    attention_mask = (
        (magnitude_cam >= threshold)
        &
        brain_mask
        &
        (magnitude_cam > 0)
    )


    return (
        attention_mask,
        threshold
    )


# ============================================================
# 3. RUN ALL TEST PATIENTS
# ============================================================

records = []


for i in tqdm(
    range(len(test_multimodal_dataset)),
    desc="Grad-CAM vs ET mask"
):

    # --------------------------------------------------------
    # Patient data
    # --------------------------------------------------------

    t1_i, t2_i, clinical_i, true_label = (
        test_multimodal_dataset[i]
    )

    patient_id = str(
        test_df.iloc[i]["PatientID"]
    )


    # --------------------------------------------------------
    # Signed Grad-CAM
    # --------------------------------------------------------

    result = generate_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(
            device
        ),

        t2_i.unsqueeze(0).to(
            device
        ),

        clinical_i.unsqueeze(0).to(
            device
        )
    )


    mag1 = result[
        "magnitude_t1"
    ]

    mag2 = result[
        "magnitude_t2"
    ]


    pred_class = result[
        "prediction"
    ]

    probability = result[
        "probability"
    ]


    # --------------------------------------------------------
    # Load expert segmentation masks
    # --------------------------------------------------------

    npz_path = os.path.join(
        PROCESSED_DIR,
        f"{patient_id}.npz"
    )


    with np.load(npz_path) as data:

        seg1 = data[
            "seg1"
        ]

        seg2 = data[
            "seg2"
        ]


    # ET = label 3
    et1 = (
        seg1 == 3
    )

    et2 = (
        seg2 == 3
    )


    # --------------------------------------------------------
    # Brain masks
    # --------------------------------------------------------

    t1_np = t1_i.numpy()
    t2_np = t2_i.numpy()


    brain1 = np.any(
        np.abs(t1_np) > 1e-6,
        axis=1
    )

    brain2 = np.any(
        np.abs(t2_np) > 1e-6,
        axis=1
    )


    # --------------------------------------------------------
    # Binary attention masks
    # --------------------------------------------------------

    attention1, threshold1 = (
        make_attention_mask(
            mag1,
            brain1,
            percentile=80
        )
    )

    attention2, threshold2 = (
        make_attention_mask(
            mag2,
            brain2,
            percentile=80
        )
    )


    # --------------------------------------------------------
    # Time1 metrics
    # --------------------------------------------------------

    (
        dice1,
        iou1,
        precision1,
        recall1
    ) = calculate_overlap_metrics(
        attention1,
        et1
    )


    # --------------------------------------------------------
    # Time2 metrics
    # --------------------------------------------------------

    (
        dice2,
        iou2,
        precision2,
        recall2
    ) = calculate_overlap_metrics(
        attention2,
        et2
    )


    # --------------------------------------------------------
    # Save patient result
    # --------------------------------------------------------

    records.append({

        "PatientID":
            patient_id,

        "TrueLabel":
            class_names[int(true_label)],

        "PredictedLabel":
            class_names[pred_class],

        "Probability":
            probability,


        # Time1
        "T1_Dice":
            dice1,

        "T1_IoU":
            iou1,

        "T1_AttentionPrecision":
            precision1,

        "T1_AttentionRecall":
            recall1,

        "T1_CAM_Threshold":
            threshold1,

        "T1_ET_Voxels":
            int(et1.sum()),

        "T1_Attention_Voxels":
            int(attention1.sum()),


        # Time2
        "T2_Dice":
            dice2,

        "T2_IoU":
            iou2,

        "T2_AttentionPrecision":
            precision2,

        "T2_AttentionRecall":
            recall2,

        "T2_CAM_Threshold":
            threshold2,

        "T2_ET_Voxels":
            int(et2.sum()),

        "T2_Attention_Voxels":
            int(attention2.sum())
    })


# ============================================================
# 4. SAVE FULL RESULTS
# ============================================================

xai_df = pd.DataFrame(
    records
)


OUTPUT_DIR = os.path.join(
    DATA_DIR,
    "research_outputs"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


xai_path = os.path.join(
    OUTPUT_DIR,
    "gradcam_et_overlap_test.csv"
)


xai_df.to_csv(
    xai_path,
    index=False
)


# ============================================================
# 5. OVERALL RESULTS
# ============================================================

print("\n======================================")
print("GRAD-CAM vs EXPERT ET MASK")
print("======================================")

print("\nTIME1")

print(
    "Mean Dice      :",
    round(
        xai_df["T1_Dice"].mean(),
        4
    )
)

print(
    "Mean IoU       :",
    round(
        xai_df["T1_IoU"].mean(),
        4
    )
)

print(
    "Mean Precision :",
    round(
        xai_df[
            "T1_AttentionPrecision"
        ].mean(),
        4
    )
)

print(
    "Mean Recall    :",
    round(
        xai_df[
            "T1_AttentionRecall"
        ].mean(),
        4
    )
)


print("\nTIME2")

print(
    "Mean Dice      :",
    round(
        xai_df["T2_Dice"].mean(),
        4
    )
)

print(
    "Mean IoU       :",
    round(
        xai_df["T2_IoU"].mean(),
        4
    )
)

print(
    "Mean Precision :",
    round(
        xai_df[
            "T2_AttentionPrecision"
        ].mean(),
        4
    )
)

print(
    "Mean Recall    :",
    round(
        xai_df[
            "T2_AttentionRecall"
        ].mean(),
        4
    )
)


print(
    "\n✅ Full XAI results saved at:"
)

print(
    xai_path
)
```

```python
# ============================================================
# STEP 19: TUMOR EVOLUTION vs ATTENTION EVOLUTION
#
# Actual tumor evolution:
#   Growth     = ET present at T2 but not T1
#   Regression = ET present at T1 but not T2
#
# Attention evolution:
#   Delta Attention = |CAM_T2| - |CAM_T1|
#
# Positive delta -> attention increased
# Negative delta -> attention decreased
#
# Temporal Alignment Score:
#   +1 = excellent directional alignment
#    0 = no clear alignment
#   -1 = opposite alignment
# ============================================================

import os
import numpy as np
import pandas as pd

from tqdm.auto import tqdm


# ------------------------------------------------------------
# Binary overlap metric
# ------------------------------------------------------------

def overlap_metrics(pred_mask, true_mask):

    pred_mask = pred_mask.astype(bool)
    true_mask = true_mask.astype(bool)

    intersection = np.logical_and(
        pred_mask,
        true_mask
    ).sum()

    union = np.logical_or(
        pred_mask,
        true_mask
    ).sum()

    pred_size = pred_mask.sum()
    true_size = true_mask.sum()


    if pred_size + true_size == 0:
        dice = np.nan
    else:
        dice = (
            2 * intersection
        ) / (
            pred_size + true_size
        )


    if union == 0:
        iou = np.nan
    else:
        iou = intersection / union


    return dice, iou


# ------------------------------------------------------------
# Fixed top-20% change mask
# ------------------------------------------------------------

def top_change_mask(
    change_map,
    brain_mask,
    mode="increase",
    percentile=80
):

    if mode == "increase":

        values_map = np.maximum(
            change_map,
            0
        )

    else:

        values_map = np.maximum(
            -change_map,
            0
        )


    values = values_map[
        brain_mask
    ]

    positive = values[
        values > 0
    ]


    if len(positive) == 0:

        return np.zeros_like(
            brain_mask,
            dtype=bool
        )


    threshold = np.percentile(
        positive,
        percentile
    )


    return (
        (values_map >= threshold)
        &
        brain_mask
        &
        (values_map > 0)
    )


# ============================================================
# ALL TEST PATIENTS
# ============================================================

temporal_records = []


for i in tqdm(
    range(len(test_multimodal_dataset)),
    desc="Temporal alignment"
):

    t1_i, t2_i, clinical_i, true_label = (
        test_multimodal_dataset[i]
    )

    patient_id = str(
        test_df.iloc[i]["PatientID"]
    )


    # --------------------------------------------------------
    # Signed Grad-CAM
    # --------------------------------------------------------

    result = generate_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(device),

        t2_i.unsqueeze(0).to(device),

        clinical_i.unsqueeze(0).to(device)
    )


    mag1 = result["magnitude_t1"]
    mag2 = result["magnitude_t2"]


    # Attention temporal change
    attention_delta = mag2 - mag1


    # --------------------------------------------------------
    # Load ET masks
    # --------------------------------------------------------

    npz_path = os.path.join(
        PROCESSED_DIR,
        f"{patient_id}.npz"
    )


    with np.load(npz_path) as data:

        seg1 = data["seg1"]
        seg2 = data["seg2"]


    et1 = (seg1 == 3)
    et2 = (seg2 == 3)


    # --------------------------------------------------------
    # Actual tumor evolution
    # --------------------------------------------------------

    growth = (
        et2 & (~et1)
    )

    regression = (
        et1 & (~et2)
    )

    stable = (
        et1 & et2
    )


    # --------------------------------------------------------
    # Brain mask
    # --------------------------------------------------------

    t1_np = t1_i.numpy()
    t2_np = t2_i.numpy()


    brain = (
        np.any(
            np.abs(t1_np) > 1e-6,
            axis=1
        )
        |
        np.any(
            np.abs(t2_np) > 1e-6,
            axis=1
        )
    )


    # --------------------------------------------------------
    # Attention increase/decrease masks
    # --------------------------------------------------------

    attention_increase = top_change_mask(
        attention_delta,
        brain,
        mode="increase",
        percentile=80
    )


    attention_decrease = top_change_mask(
        attention_delta,
        brain,
        mode="decrease",
        percentile=80
    )


    # --------------------------------------------------------
    # Growth overlap
    # --------------------------------------------------------

    growth_dice, growth_iou = overlap_metrics(
        attention_increase,
        growth
    )


    # --------------------------------------------------------
    # Regression overlap
    # --------------------------------------------------------

    regression_dice, regression_iou = overlap_metrics(
        attention_decrease,
        regression
    )


    # ========================================================
    # TEMPORAL ALIGNMENT SCORE
    #
    # On actual changed-tumor voxels only:
    #
    # Growth voxel     expects positive attention change
    # Regression voxel expects negative attention change
    #
    # Range approximately [-1, +1]
    # ========================================================

    change_region = (
        growth | regression
    )


    if change_region.sum() > 0:

        target_direction = np.zeros_like(
            attention_delta,
            dtype=np.float32
        )

        target_direction[growth] = 1.0

        target_direction[regression] = -1.0


        max_delta = np.max(
            np.abs(attention_delta)
        )


        if max_delta > 0:

            normalized_delta = (
                attention_delta /
                max_delta
            )

            temporal_alignment = np.mean(
                target_direction[change_region]
                *
                normalized_delta[change_region]
            )

        else:

            temporal_alignment = 0.0

    else:

        temporal_alignment = np.nan


    # --------------------------------------------------------
    # Mean attention change inside growth/regression
    # --------------------------------------------------------

    growth_attention_change = (
        np.mean(attention_delta[growth])
        if growth.sum() > 0
        else np.nan
    )


    regression_attention_change = (
        np.mean(attention_delta[regression])
        if regression.sum() > 0
        else np.nan
    )


    # --------------------------------------------------------
    # Record
    # --------------------------------------------------------

    temporal_records.append({

        "PatientID":
            patient_id,

        "TrueLabel":
            class_names[int(true_label)],


        "T1_ET_Voxels":
            int(et1.sum()),

        "T2_ET_Voxels":
            int(et2.sum()),


        "Growth_Voxels":
            int(growth.sum()),

        "Regression_Voxels":
            int(regression.sum()),

        "Stable_Voxels":
            int(stable.sum()),


        "Growth_Dice":
            growth_dice,

        "Growth_IoU":
            growth_iou,


        "Regression_Dice":
            regression_dice,

        "Regression_IoU":
            regression_iou,


        "Mean_AttentionChange_Growth":
            growth_attention_change,

        "Mean_AttentionChange_Regression":
            regression_attention_change,


        "TemporalAlignmentScore":
            temporal_alignment
    })


# ============================================================
# SAVE RESULTS
# ============================================================

temporal_df = pd.DataFrame(
    temporal_records
)


temporal_path = os.path.join(
    OUTPUT_DIR,
    "temporal_attention_tumor_alignment_test.csv"
)


temporal_df.to_csv(
    temporal_path,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n======================================")
print("TEMPORAL ALIGNMENT RESULTS")
print("======================================")


print(
    "Mean Growth Dice:",
    round(
        temporal_df[
            "Growth_Dice"
        ].mean(),
        4
    )
)

print(
    "Mean Growth IoU:",
    round(
        temporal_df[
            "Growth_IoU"
        ].mean(),
        4
    )
)


print(
    "Mean Regression Dice:",
    round(
        temporal_df[
            "Regression_Dice"
        ].mean(),
        4
    )
)

print(
    "Mean Regression IoU:",
    round(
        temporal_df[
            "Regression_IoU"
        ].mean(),
        4
    )
)


print(
    "Mean Temporal Alignment Score:",
    round(
        temporal_df[
            "TemporalAlignmentScore"
        ].mean(),
        4
    )
)


print(
    "\nPositive temporal alignment patients:",
    int(
        (
            temporal_df[
                "TemporalAlignmentScore"
            ] > 0
        ).sum()
    ),
    "/",
    temporal_df[
        "TemporalAlignmentScore"
    ].notna().sum()
)


print(
    "\n✅ Temporal results saved:"
)

print(
    temporal_path
)
```

```python
# ============================================================
# STEP 20.1: MULTI-LAYER SIGNED GRAD-CAM
#
# Purpose:
#   ResNet layer3 + layer4 Grad-CAM combine করা।
#
# Why:
#   layer4 = coarse 7x7
#   layer3 = finer 14x14
#
#   Equal-weight combination:
#       50% layer3 + 50% layer4
#
# IMPORTANT:
#   - Signed contribution preserved
#   - No tumor mask used
#   - No threshold tuning
# ============================================================

import numpy as np
import torch
import torch.nn.functional as F


def generate_multilayer_signed_gradcam(
    model,
    time1,
    time2,
    clinical
):

    model.eval()
    model.zero_grad(set_to_none=True)

    time1 = time1.clone().detach().requires_grad_(True)
    time2 = time2.clone().detach().requires_grad_(True)

    captures = {
        "layer3": [],
        "layer4": []
    }


    # --------------------------------------------------------
    # Hook creator
    # --------------------------------------------------------

    def make_hook(layer_name):

        def hook(module, inputs, output):

            record = {
                "activation": output.detach()
            }

            def save_gradient(grad):

                record["gradient"] = grad.detach()

            output.register_hook(
                save_gradient
            )

            captures[layer_name].append(
                record
            )

        return hook


    # --------------------------------------------------------
    # Register layer3 + layer4 hooks
    # --------------------------------------------------------

    h3 = model.encoder.layer3[-1].register_forward_hook(
        make_hook("layer3")
    )

    h4 = model.encoder.layer4[-1].register_forward_hook(
        make_hook("layer4")
    )


    # --------------------------------------------------------
    # Forward
    # --------------------------------------------------------

    logits = model(
        time1,
        time2,
        clinical
    )

    pred_class = int(
        logits.argmax(dim=1).item()
    )

    probability = float(
        torch.softmax(
            logits,
            dim=1
        )[0, pred_class].item()
    )


    # --------------------------------------------------------
    # Backward
    # --------------------------------------------------------

    logits[0, pred_class].backward()

    h3.remove()
    h4.remove()


    # --------------------------------------------------------
    # Build signed CAM
    # --------------------------------------------------------

    def build_cam(record):

        activation = record["activation"]
        gradient = record["gradient"]

        weights = gradient.mean(
            dim=(2, 3),
            keepdim=True
        )

        cam = (
            weights * activation
        ).sum(dim=1)

        cam = F.interpolate(
            cam.unsqueeze(1),
            size=(224, 224),
            mode="bilinear",
            align_corners=False
        ).squeeze(1)

        return (
            cam.detach()
            .cpu()
            .numpy()
        )


    # --------------------------------------------------------
    # Shared encoder called:
    # index 0 = Time1
    # index 1 = Time2
    # --------------------------------------------------------

    l3_t1 = build_cam(
        captures["layer3"][0]
    )

    l3_t2 = build_cam(
        captures["layer3"][1]
    )

    l4_t1 = build_cam(
        captures["layer4"][0]
    )

    l4_t2 = build_cam(
        captures["layer4"][1]
    )


    # --------------------------------------------------------
    # Brain mask
    # --------------------------------------------------------

    t1_np = (
        time1[0]
        .detach()
        .cpu()
        .numpy()
    )

    t2_np = (
        time2[0]
        .detach()
        .cpu()
        .numpy()
    )

    brain = (
        np.any(
            np.abs(t1_np) > 1e-6,
            axis=1
        )
        |
        np.any(
            np.abs(t2_np) > 1e-6,
            axis=1
        )
    )


    l3_t1 *= brain
    l3_t2 *= brain

    l4_t1 *= brain
    l4_t2 *= brain


    # --------------------------------------------------------
    # Normalize EACH layer jointly across Time1 + Time2
    # --------------------------------------------------------

    def joint_normalize(cam1, cam2):

        maximum = max(
            np.max(np.abs(cam1)),
            np.max(np.abs(cam2)),
            1e-12
        )

        return (
            cam1 / maximum,
            cam2 / maximum
        )


    l3_t1, l3_t2 = joint_normalize(
        l3_t1,
        l3_t2
    )

    l4_t1, l4_t2 = joint_normalize(
        l4_t1,
        l4_t2
    )


    # --------------------------------------------------------
    # Equal-weight multi-layer fusion
    # --------------------------------------------------------

    signed_t1 = (
        0.5 * l3_t1
        +
        0.5 * l4_t1
    )

    signed_t2 = (
        0.5 * l3_t2
        +
        0.5 * l4_t2
    )


    # Final joint normalization
    final_max = max(
        np.max(np.abs(signed_t1)),
        np.max(np.abs(signed_t2)),
        1e-12
    )

    signed_t1 = signed_t1 / final_max
    signed_t2 = signed_t2 / final_max


    magnitude_t1 = np.abs(
        signed_t1
    )

    magnitude_t2 = np.abs(
        signed_t2
    )


    return {
        "signed_t1": signed_t1,
        "signed_t2": signed_t2,

        "magnitude_t1": magnitude_t1,
        "magnitude_t2": magnitude_t2,

        "prediction": pred_class,
        "probability": probability
    }


# ============================================================
# QC ON ALL TEST PATIENTS
# ============================================================

zero_multi_t1 = []
zero_multi_t2 = []


for i in range(
    len(test_multimodal_dataset)
):

    t1_i, t2_i, clinical_i, _ = (
        test_multimodal_dataset[i]
    )

    pid = str(
        test_df.iloc[i]["PatientID"]
    )


    result = generate_multilayer_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(device),

        t2_i.unsqueeze(0).to(device),

        clinical_i.unsqueeze(0).to(device)
    )


    if result["magnitude_t1"].max() <= 1e-12:
        zero_multi_t1.append(pid)

    if result["magnitude_t2"].max() <= 1e-12:
        zero_multi_t2.append(pid)


print("======================================")
print("MULTI-LAYER SIGNED GRAD-CAM QC")
print("======================================")

print(
    "Total test patients:",
    len(test_multimodal_dataset)
)

print(
    "Zero Time1 CAM:",
    len(zero_multi_t1)
)

print(
    "Zero Time2 CAM:",
    len(zero_multi_t2)
)

print(
    "\nTime1 zero patients:",
    zero_multi_t1
)

print(
    "Time2 zero patients:",
    zero_multi_t2
)
```

```python
# ============================================================
# STEP 20.3: MULTI-LAYER TEMPORAL ALIGNMENT
#
# Purpose:
#   Multi-layer Signed Grad-CAM দিয়ে
#   actual ET growth/regression-এর সাথে attention change
#   compare করা।
#
# Same methodology as previous temporal experiment:
#
#   Growth     = ET2 & ~ET1
#   Regression = ET1 & ~ET2
#
#   Attention change =
#       |CAM_Time2| - |CAM_Time1|
#
# Output:
#   - Growth Dice / IoU
#   - Regression Dice / IoU
#   - Temporal Alignment Score
#   - Single-layer vs Multi-layer comparison
# ============================================================

import os
import numpy as np
import pandas as pd

from tqdm.auto import tqdm


multi_temporal_records = []


for i in tqdm(
    range(len(test_multimodal_dataset)),
    desc="Multi-layer temporal alignment"
):

    # --------------------------------------------------------
    # 1. Patient data
    # --------------------------------------------------------

    t1_i, t2_i, clinical_i, true_label = (
        test_multimodal_dataset[i]
    )

    patient_id = str(
        test_df.iloc[i]["PatientID"]
    )


    # --------------------------------------------------------
    # 2. Multi-layer Signed Grad-CAM
    # --------------------------------------------------------

    result = generate_multilayer_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(device),

        t2_i.unsqueeze(0).to(device),

        clinical_i.unsqueeze(0).to(device)
    )


    mag1 = result["magnitude_t1"]
    mag2 = result["magnitude_t2"]


    # Longitudinal attention change
    attention_delta = (
        mag2 - mag1
    )


    # --------------------------------------------------------
    # 3. Load segmentation masks
    # --------------------------------------------------------

    npz_path = os.path.join(
        PROCESSED_DIR,
        f"{patient_id}.npz"
    )


    with np.load(npz_path) as data:

        seg1 = data["seg1"]
        seg2 = data["seg2"]


    # ET label = 3
    et1 = (seg1 == 3)
    et2 = (seg2 == 3)


    # --------------------------------------------------------
    # 4. Actual tumor evolution
    # --------------------------------------------------------

    growth = (
        et2 & (~et1)
    )

    regression = (
        et1 & (~et2)
    )

    stable = (
        et1 & et2
    )


    # --------------------------------------------------------
    # 5. Brain mask
    # --------------------------------------------------------

    t1_np = t1_i.numpy()
    t2_np = t2_i.numpy()


    brain = (
        np.any(
            np.abs(t1_np) > 1e-6,
            axis=1
        )
        |
        np.any(
            np.abs(t2_np) > 1e-6,
            axis=1
        )
    )


    # --------------------------------------------------------
    # 6. Attention increase / decrease
    #
    # Same top-20% rule used before
    # --------------------------------------------------------

    attention_increase = top_change_mask(
        attention_delta,
        brain,
        mode="increase",
        percentile=80
    )


    attention_decrease = top_change_mask(
        attention_delta,
        brain,
        mode="decrease",
        percentile=80
    )


    # --------------------------------------------------------
    # 7. Spatial overlap
    # --------------------------------------------------------

    growth_dice, growth_iou = overlap_metrics(
        attention_increase,
        growth
    )


    regression_dice, regression_iou = overlap_metrics(
        attention_decrease,
        regression
    )


    # --------------------------------------------------------
    # 8. Temporal Alignment Score
    #
    # Growth region expects positive CAM change.
    # Regression region expects negative CAM change.
    # --------------------------------------------------------

    change_region = (
        growth | regression
    )


    if change_region.sum() > 0:

        target_direction = np.zeros_like(
            attention_delta,
            dtype=np.float32
        )

        target_direction[growth] = 1.0
        target_direction[regression] = -1.0


        max_delta = np.max(
            np.abs(attention_delta)
        )


        if max_delta > 0:

            normalized_delta = (
                attention_delta /
                max_delta
            )

            temporal_alignment = np.mean(
                target_direction[change_region]
                *
                normalized_delta[change_region]
            )

        else:

            temporal_alignment = 0.0


    else:

        temporal_alignment = np.nan


    # --------------------------------------------------------
    # 9. Mean CAM change inside actual changed regions
    # --------------------------------------------------------

    growth_attention_change = (
        np.mean(attention_delta[growth])
        if growth.sum() > 0
        else np.nan
    )


    regression_attention_change = (
        np.mean(attention_delta[regression])
        if regression.sum() > 0
        else np.nan
    )


    # --------------------------------------------------------
    # 10. Save record
    # --------------------------------------------------------

    multi_temporal_records.append({

        "PatientID":
            patient_id,

        "TrueLabel":
            class_names[int(true_label)],

        "Growth_Voxels":
            int(growth.sum()),

        "Regression_Voxels":
            int(regression.sum()),

        "Stable_Voxels":
            int(stable.sum()),

        "Growth_Dice":
            growth_dice,

        "Growth_IoU":
            growth_iou,

        "Regression_Dice":
            regression_dice,

        "Regression_IoU":
            regression_iou,

        "Mean_AttentionChange_Growth":
            growth_attention_change,

        "Mean_AttentionChange_Regression":
            regression_attention_change,

        "TemporalAlignmentScore":
            temporal_alignment
    })


# ============================================================
# 11. SAVE RESULTS
# ============================================================

multi_temporal_df = pd.DataFrame(
    multi_temporal_records
)


multi_temporal_path = os.path.join(
    OUTPUT_DIR,
    "multilayer_temporal_alignment_test.csv"
)


multi_temporal_df.to_csv(
    multi_temporal_path,
    index=False
)


# ============================================================
# 12. RESULTS
# ============================================================

print("\n======================================")
print("MULTI-LAYER TEMPORAL ALIGNMENT")
print("======================================")


print(
    "Mean Growth Dice:",
    round(
        multi_temporal_df[
            "Growth_Dice"
        ].mean(),
        4
    )
)

print(
    "Mean Growth IoU:",
    round(
        multi_temporal_df[
            "Growth_IoU"
        ].mean(),
        4
    )
)


print(
    "Mean Regression Dice:",
    round(
        multi_temporal_df[
            "Regression_Dice"
        ].mean(),
        4
    )
)

print(
    "Mean Regression IoU:",
    round(
        multi_temporal_df[
            "Regression_IoU"
        ].mean(),
        4
    )
)


multi_alignment = (
    multi_temporal_df[
        "TemporalAlignmentScore"
    ].mean()
)


print(
    "Mean Temporal Alignment Score:",
    round(
        multi_alignment,
        4
    )
)


valid_multi = multi_temporal_df[
    "TemporalAlignmentScore"
].notna()


print(
    "Positive alignment patients:",
    int(
        (
            multi_temporal_df.loc[
                valid_multi,
                "TemporalAlignmentScore"
            ] > 0
        ).sum()
    ),
    "/",
    int(valid_multi.sum())
)


# ============================================================
# 13. SINGLE vs MULTI COMPARISON
# ============================================================

print("\n======================================")
print("SINGLE vs MULTI TEMPORAL COMPARISON")
print("======================================")


print(
    "Single-layer alignment:",
    round(
        temporal_df[
            "TemporalAlignmentScore"
        ].mean(),
        4
    )
)

print(
    "Multi-layer alignment :",
    round(
        multi_alignment,
        4
    )
)


print(
    "\nSingle Growth Dice:",
    round(
        temporal_df[
            "Growth_Dice"
        ].mean(),
        4
    )
)

print(
    "Multi Growth Dice :",
    round(
        multi_temporal_df[
            "Growth_Dice"
        ].mean(),
        4
    )
)


print(
    "\nSingle Regression Dice:",
    round(
        temporal_df[
            "Regression_Dice"
        ].mean(),
        4
    )
)

print(
    "Multi Regression Dice :",
    round(
        multi_temporal_df[
            "Regression_Dice"
        ].mean(),
        4
    )
)


print(
    "\n✅ Results saved:"
)

print(
    multi_temporal_path
)
```

```python
# ============================================================
# STEP 21.1: TEMPORAL ALIGNMENT STATISTICAL ANALYSIS
#
# Purpose:
#   Single-layer এবং Multi-layer temporal alignment
#   statistically zero থেকে different কিনা পরীক্ষা করা।
#
# Tests:
#   1. Mean
#   2. Median
#   3. Bootstrap 95% CI
#   4. Wilcoxon signed-rank vs zero
#   5. Paired Wilcoxon: Single vs Multi
#
# p < 0.05 = statistically significant
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon


# ------------------------------------------------------------
# 1. Merge same patients from both methods
# ------------------------------------------------------------

stats_df = temporal_df[
    ["PatientID", "TemporalAlignmentScore"]
].rename(
    columns={
        "TemporalAlignmentScore": "Single"
    }
).merge(

    multi_temporal_df[
        ["PatientID", "TemporalAlignmentScore"]
    ].rename(
        columns={
            "TemporalAlignmentScore": "Multi"
        }
    ),

    on="PatientID",
    how="inner"
)


# Keep patients with valid scores in BOTH methods
stats_df = stats_df.dropna(
    subset=["Single", "Multi"]
).reset_index(drop=True)


single_scores = stats_df[
    "Single"
].values

multi_scores = stats_df[
    "Multi"
].values


print(
    "Valid paired patients:",
    len(stats_df)
)


# ============================================================
# 2. Bootstrap 95% CI function
# ============================================================

def bootstrap_ci(
    values,
    n_bootstrap=10000,
    seed=42
):

    rng = np.random.default_rng(seed)

    bootstrap_means = []

    for _ in range(n_bootstrap):

        sample = rng.choice(
            values,
            size=len(values),
            replace=True
        )

        bootstrap_means.append(
            np.mean(sample)
        )


    lower = np.percentile(
        bootstrap_means,
        2.5
    )

    upper = np.percentile(
        bootstrap_means,
        97.5
    )

    return lower, upper


# ============================================================
# 3. SINGLE-LAYER STATISTICS
# ============================================================

single_mean = np.mean(
    single_scores
)

single_median = np.median(
    single_scores
)

single_ci = bootstrap_ci(
    single_scores
)

single_stat, single_p = wilcoxon(
    single_scores,
    alternative="two-sided"
)


# ============================================================
# 4. MULTI-LAYER STATISTICS
# ============================================================

multi_mean = np.mean(
    multi_scores
)

multi_median = np.median(
    multi_scores
)

multi_ci = bootstrap_ci(
    multi_scores
)

multi_stat, multi_p = wilcoxon(
    multi_scores,
    alternative="two-sided"
)


# ============================================================
# 5. SINGLE vs MULTI PAIRED COMPARISON
# ============================================================

compare_stat, compare_p = wilcoxon(
    single_scores,
    multi_scores,
    alternative="two-sided"
)


# ============================================================
# 6. PRINT RESULTS
# ============================================================

print("\n======================================")
print("SINGLE-LAYER TEMPORAL STATISTICS")
print("======================================")

print(
    "Mean       :",
    round(single_mean, 6)
)

print(
    "Median     :",
    round(single_median, 6)
)

print(
    "95% CI     :",
    (
        round(single_ci[0], 6),
        round(single_ci[1], 6)
    )
)

print(
    "Wilcoxon p :",
    round(float(single_p), 6)
)


print("\n======================================")
print("MULTI-LAYER TEMPORAL STATISTICS")
print("======================================")

print(
    "Mean       :",
    round(multi_mean, 6)
)

print(
    "Median     :",
    round(multi_median, 6)
)

print(
    "95% CI     :",
    (
        round(multi_ci[0], 6),
        round(multi_ci[1], 6)
    )
)

print(
    "Wilcoxon p :",
    round(float(multi_p), 6)
)


print("\n======================================")
print("SINGLE vs MULTI")
print("======================================")

print(
    "Paired Wilcoxon p:",
    round(float(compare_p), 6)
)


# ============================================================
# 7. Simple interpretation
# ============================================================

print("\nINTERPRETATION")

if single_p < 0.05:

    print(
        "✅ Single-layer temporal alignment is "
        "significantly different from zero."
    )

else:

    print(
        "⚠️ Single-layer temporal alignment is "
        "NOT significantly different from zero."
    )


if multi_p < 0.05:

    print(
        "✅ Multi-layer temporal alignment is "
        "significantly different from zero."
    )

else:

    print(
        "⚠️ Multi-layer temporal alignment is "
        "NOT significantly different from zero."
    )


if compare_p < 0.05:

    print(
        "✅ Single and Multi-layer alignment "
        "are significantly different."
    )

else:

    print(
        "ℹ️ No significant difference between "
        "Single and Multi-layer temporal alignment."
    )
```

```python
# ============================================================
# STEP 21.2: TUMOR CENTROID SHIFT vs ATTENTION CENTROID SHIFT
#
# Purpose:
#   Time1 -> Time2 actual ET tumor centroid কতটা সরে গেছে
#   এবং Grad-CAM attention centroid কতটা সরে গেছে,
#   তাদের relationship measure করা।
#
# Metrics:
#   - Tumor shift distance
#   - Attention shift distance
#   - Direction cosine similarity
#   - Pearson correlation
#   - Spearman correlation
#
# Coordinates normalized to 0-1 so z/y/x scale comparable.
# ============================================================

import os
import numpy as np
import pandas as pd

from scipy.stats import pearsonr, spearmanr
from tqdm.auto import tqdm


# ------------------------------------------------------------
# 1. Binary tumor centroid
# ------------------------------------------------------------

def binary_centroid(mask):

    mask = mask.astype(bool)

    if mask.sum() == 0:
        return None

    coords = np.argwhere(mask)

    centroid = coords.mean(axis=0)

    shape_scale = np.array(mask.shape) - 1

    return centroid / shape_scale


# ------------------------------------------------------------
# 2. Attention weighted centroid
# ------------------------------------------------------------

def weighted_centroid(attention):

    attention = np.asarray(
        attention,
        dtype=np.float64
    )

    total = attention.sum()

    if total <= 1e-12:
        return None


    z, y, x = np.indices(
        attention.shape
    )


    centroid = np.array([

        (z * attention).sum() / total,

        (y * attention).sum() / total,

        (x * attention).sum() / total

    ])


    shape_scale = (
        np.array(attention.shape) - 1
    )


    return centroid / shape_scale


# ============================================================
# 3. ALL TEST PATIENTS
# ============================================================

shift_records = []


for i in tqdm(
    range(len(test_multimodal_dataset)),
    desc="Tumor vs attention shift"
):

    t1_i, t2_i, clinical_i, true_label = (
        test_multimodal_dataset[i]
    )

    patient_id = str(
        test_df.iloc[i]["PatientID"]
    )


    # --------------------------------------------------------
    # Signed Grad-CAM magnitude
    # --------------------------------------------------------

    result = generate_signed_gradcam(

        model_improved,

        t1_i.unsqueeze(0).to(device),

        t2_i.unsqueeze(0).to(device),

        clinical_i.unsqueeze(0).to(device)
    )


    attention1 = result[
        "magnitude_t1"
    ]

    attention2 = result[
        "magnitude_t2"
    ]


    # --------------------------------------------------------
    # Load ET masks
    # --------------------------------------------------------

    npz_path = os.path.join(
        PROCESSED_DIR,
        f"{patient_id}.npz"
    )


    with np.load(npz_path) as data:

        seg1 = data["seg1"]
        seg2 = data["seg2"]


    et1 = (seg1 == 3)
    et2 = (seg2 == 3)


    # --------------------------------------------------------
    # Centroids
    # --------------------------------------------------------

    tumor_c1 = binary_centroid(
        et1
    )

    tumor_c2 = binary_centroid(
        et2
    )


    attention_c1 = weighted_centroid(
        attention1
    )

    attention_c2 = weighted_centroid(
        attention2
    )


    # Need tumor at BOTH timepoints
    if (
        tumor_c1 is None
        or
        tumor_c2 is None
        or
        attention_c1 is None
        or
        attention_c2 is None
    ):

        continue


    # --------------------------------------------------------
    # Shift vectors
    # --------------------------------------------------------

    tumor_vector = (
        tumor_c2 - tumor_c1
    )

    attention_vector = (
        attention_c2 - attention_c1
    )


    tumor_distance = np.linalg.norm(
        tumor_vector
    )

    attention_distance = np.linalg.norm(
        attention_vector
    )


    # --------------------------------------------------------
    # Direction similarity
    #
    # +1 = same direction
    #  0 = unrelated direction
    # -1 = opposite direction
    # --------------------------------------------------------

    if (
        tumor_distance > 1e-12
        and
        attention_distance > 1e-12
    ):

        cosine_similarity = (
            np.dot(
                tumor_vector,
                attention_vector
            )
            /
            (
                tumor_distance
                *
                attention_distance
            )
        )

    else:

        cosine_similarity = np.nan


    shift_records.append({

        "PatientID":
            patient_id,

        "TrueLabel":
            class_names[int(true_label)],

        "TumorShiftDistance":
            tumor_distance,

        "AttentionShiftDistance":
            attention_distance,

        "ShiftDirectionCosine":
            cosine_similarity
    })


# ============================================================
# 4. DATAFRAME
# ============================================================

shift_df = pd.DataFrame(
    shift_records
)


# Remove invalid values
corr_df = shift_df.dropna(
    subset=[
        "TumorShiftDistance",
        "AttentionShiftDistance"
    ]
)


# ============================================================
# 5. PEARSON + SPEARMAN
# ============================================================

pearson_r, pearson_p = pearsonr(

    corr_df[
        "TumorShiftDistance"
    ],

    corr_df[
        "AttentionShiftDistance"
    ]
)


spearman_r, spearman_p = spearmanr(

    corr_df[
        "TumorShiftDistance"
    ],

    corr_df[
        "AttentionShiftDistance"
    ]
)


# ============================================================
# 6. DIRECTION SIMILARITY
# ============================================================

valid_cosine = shift_df[
    "ShiftDirectionCosine"
].dropna()


mean_cosine = valid_cosine.mean()

median_cosine = valid_cosine.median()


# ============================================================
# 7. SAVE
# ============================================================

shift_path = os.path.join(
    OUTPUT_DIR,
    "tumor_attention_centroid_shift.csv"
)


shift_df.to_csv(
    shift_path,
    index=False
)


# ============================================================
# 8. RESULTS
# ============================================================

print("\n======================================")
print("TUMOR vs ATTENTION CENTROID SHIFT")
print("======================================")

print(
    "Valid patients:",
    len(corr_df)
)


print("\nSHIFT MAGNITUDE CORRELATION")

print(
    "Pearson r :",
    round(float(pearson_r), 4)
)

print(
    "Pearson p :",
    round(float(pearson_p), 6)
)

print(
    "Spearman r:",
    round(float(spearman_r), 4)
)

print(
    "Spearman p:",
    round(float(spearman_p), 6)
)


print("\nSHIFT DIRECTION")

print(
    "Mean cosine similarity:",
    round(float(mean_cosine), 4)
)

print(
    "Median cosine similarity:",
    round(float(median_cosine), 4)
)


print(
    "\n✅ Results saved:"
)

print(
    shift_path
)
```

```python
# ============================================================
# STEP 10: SINGLE-TIME2 CNN BASELINE
#
# Purpose:
#   শুধু Time2 FLAIR + T1CE MRI ব্যবহার করে progression predict করা।
#
# Fair comparison:
#   Time1 baseline-এর SAME architecture/training strategy ব্যবহার করছি।
#
# Output:
#   - Best Validation Macro-F1
#   - best_time2_cnn.pt
#   - time2_cnn_history.csv
#
# Test set এখানে ব্যবহার করা হচ্ছে না.
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, f1_score


# ============================================================
# 1. TIME2 DATASET
# ============================================================

class Time2MRIDataset(Dataset):

    def __init__(self, dataframe, processed_dir):

        self.df = dataframe.reset_index(drop=True)
        self.processed_dir = processed_dir


    def __len__(self):

        return len(self.df)


    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        patient_id = str(
            row["PatientID"]
        )

        path = os.path.join(
            self.processed_dir,
            f"{patient_id}.npz"
        )

        with np.load(path) as data:

            image = data["time2"].astype(
                np.float32
            )

        label = label_map[
            row["Label"]
        ]

        return (
            torch.from_numpy(image),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )


# ============================================================
# 2. DATALOADERS
# ============================================================

train_time2_dataset = Time2MRIDataset(
    train_df,
    PROCESSED_DIR
)

val_time2_dataset = Time2MRIDataset(
    val_df,
    PROCESSED_DIR
)


train_loader_time2 = DataLoader(
    train_time2_dataset,
    batch_size=1,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader_time2 = DataLoader(
    val_time2_dataset,
    batch_size=1,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


print(
    "Training patients:",
    len(train_time2_dataset)
)

print(
    "Validation patients:",
    len(val_time2_dataset)
)


# ============================================================
# 3. CREATE SAME CNN AS TIME1 BASELINE
# ============================================================

model_time2 = Time1PatientCNN(
    num_classes=3
).to(device)


# Freeze early encoder
for param in model_time2.encoder.parameters():
    param.requires_grad = False

# Fine-tune layer4
for param in model_time2.encoder.layer4.parameters():
    param.requires_grad = True

# Classifier trainable
for param in model_time2.classifier.parameters():
    param.requires_grad = True


# ============================================================
# 4. LOSS + OPTIMIZER
# ============================================================

weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=weights_tensor
)


optimizer = torch.optim.AdamW(
    [
        p for p in model_time2.parameters()
        if p.requires_grad
    ],
    lr=1e-4,
    weight_decay=1e-4
)


scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)


scaler = torch.amp.GradScaler(
    "cuda"
)


# ============================================================
# 5. SETTINGS
# ============================================================

EPOCHS = 15
PATIENCE = 5

best_f1 = -1.0
no_improve = 0

history = []


MODEL_DIR = os.path.join(
    DATA_DIR,
    "research_models"
)

best_time2_path = os.path.join(
    MODEL_DIR,
    "best_time2_cnn.pt"
)

history_path = os.path.join(
    MODEL_DIR,
    "time2_cnn_history.csv"
)


# ============================================================
# 6. TRAINING
# ============================================================

for epoch in range(EPOCHS):

    # ---------------- TRAIN ----------------

    model_time2.train()

    train_losses = []
    train_true = []
    train_pred = []


    for images, labels in train_loader_time2:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.amp.autocast(
            device_type="cuda"
        ):

            outputs = model_time2(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()


        train_losses.append(
            loss.item()
        )

        preds = outputs.argmax(
            dim=1
        )

        train_true.extend(
            labels.cpu().numpy()
        )

        train_pred.extend(
            preds.detach().cpu().numpy()
        )


    train_acc = accuracy_score(
        train_true,
        train_pred
    )

    train_f1 = f1_score(
        train_true,
        train_pred,
        average="macro",
        zero_division=0
    )


    # ---------------- VALIDATION ----------------

    model_time2.eval()

    val_true = []
    val_pred = []


    with torch.no_grad():

        for images, labels in val_loader_time2:

            images = images.to(device)
            labels = labels.to(device)


            with torch.amp.autocast(
                device_type="cuda"
            ):

                outputs = model_time2(
                    images
                )


            preds = outputs.argmax(
                dim=1
            )

            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                preds.cpu().numpy()
            )


    val_acc = accuracy_score(
        val_true,
        val_pred
    )

    val_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )


    scheduler.step(
        val_f1
    )


    history.append({
        "epoch": epoch + 1,
        "train_accuracy": train_acc,
        "train_macro_f1": train_f1,
        "val_accuracy": val_acc,
        "val_macro_f1": val_f1
    })


    # ---------------- SAVE BEST ----------------

    if val_f1 > best_f1:

        best_f1 = val_f1
        no_improve = 0


        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict":
                    model_time2.state_dict(),
                "val_accuracy":
                    val_acc,
                "val_macro_f1":
                    val_f1
            },
            best_time2_path
        )

        status = " ⭐ BEST SAVED"

    else:

        no_improve += 1
        status = ""


    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Train F1: {train_f1:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_f1:.4f}"
        f"{status}"
    )


    if no_improve >= PATIENCE:

        print(
            "\nEarly stopping triggered."
        )

        break


# ============================================================
# 7. SAVE HISTORY
# ============================================================

pd.DataFrame(
    history
).to_csv(
    history_path,
    index=False
)


print("\n======================================")
print("TIME2 CNN BASELINE COMPLETE")
print("======================================")

print(
    "Best Validation Macro-F1:",
    round(best_f1, 4)
)

print(
    "Best model:",
    best_time2_path
)

print(
    "History:",
    history_path
)
```

```python
# ============================================================
# DRIVE-CONNECTED GLIOMA DIGITAL TWIN WEBSITE
#
# কী করছে:
#   - File upload বাদ
#   - Patient ID dropdown
#   - MRI সরাসরি Google Drive থেকে load
#   - Age / Sex / Diagnosis automatically নেয়
#   - Prediction + probability
#   - Digital Twin latent state
#   - Time1 / Time2 Grad-CAM
#   - Tumor evolution
#   - শেষে clickable website link
#
# No account / API key required.
# ============================================================

import os
import re
import time
import textwrap
import subprocess
import pandas as pd


DATA_DIR = "/content/drive/MyDrive/clean_dataset"

APP_DIR = os.path.join(
    DATA_DIR,
    "digital_twin_app"
)

PROCESSED_DIR = os.path.join(
    DATA_DIR,
    "processed_final"
)

os.makedirs(
    APP_DIR,
    exist_ok=True
)


# ============================================================
# 1. SAVE PATIENT METADATA FOR WEBSITE
# ============================================================

metadata_columns = [
    "PatientID",
    "Age",
    "Sex",
    "Diagnosis",
    "Label"
]

patient_metadata = (
    split_df[metadata_columns]
    .copy()
    .drop_duplicates(
        subset=["PatientID"]
    )
)

patient_metadata[
    "PatientID"
] = patient_metadata[
    "PatientID"
].astype(str)


metadata_path = os.path.join(
    APP_DIR,
    "patient_metadata.csv"
)

patient_metadata.to_csv(
    metadata_path,
    index=False
)

print(
    "✅ Patient metadata saved:",
    len(patient_metadata),
    "patients"
)


# ============================================================
# 2. NEW APP.PY
# ============================================================

app_code = r'''
import os
import joblib
import numpy as np
import pandas as pd
import streamlit as st
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

from torchvision.models import resnet18


# ============================================================
# SETTINGS
# ============================================================

st.set_page_config(
    page_title="Glioma Digital Twin",
    page_icon="🧠",
    layout="wide"
)

CLASS_NAMES = [
    "Unchanged",
    "Increased",
    "Decreased"
]

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

BASE_DIR = os.path.dirname(
    os.path.abspath(__file__)
)

MODEL_PATH = os.path.join(
    BASE_DIR,
    "best_model.pt"
)

PREPROCESSOR_PATH = os.path.join(
    BASE_DIR,
    "clinical_preprocessor.joblib"
)

METADATA_PATH = os.path.join(
    BASE_DIR,
    "patient_metadata.csv"
)

PROCESSED_DIR = (
    "/content/drive/MyDrive/clean_dataset/processed_final"
)


# ============================================================
# MODEL
# ============================================================

class ImprovedSiameseClinicalCNN(nn.Module):

    def __init__(
        self,
        clinical_dim=11,
        num_classes=3
    ):

        super().__init__()

        backbone = resnet18(
            weights=None
        )

        old_conv = backbone.conv1

        backbone.conv1 = nn.Conv2d(
            2,
            old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding,
            bias=False
        )

        self.feature_dim = (
            backbone.fc.in_features
        )

        backbone.fc = nn.Identity()

        self.encoder = backbone


        self.slice_attention = nn.Sequential(
            nn.Linear(
                self.feature_dim,
                128
            ),
            nn.Tanh(),
            nn.Linear(
                128,
                1
            )
        )


        self.mri_projection = nn.Sequential(
            nn.Linear(
                self.feature_dim * 3,
                256
            ),
            nn.LayerNorm(256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(
                256,
                128
            ),
            nn.ReLU()
        )


        self.clinical_projection = nn.Sequential(
            nn.Linear(
                clinical_dim,
                64
            ),
            nn.LayerNorm(64),
            nn.ReLU(),
            nn.Dropout(0.2)
        )


        self.fusion = nn.Sequential(
            nn.Linear(
                192,
                128
            ),
            nn.LayerNorm(128),
            nn.ReLU(),
            nn.Dropout(0.4)
        )


        self.classifier = nn.Linear(
            128,
            num_classes
        )


    def encode_patient(
        self,
        x
    ):

        B, S, C, H, W = x.shape

        x = x.reshape(
            B * S,
            C,
            H,
            W
        )

        features = self.encoder(
            x
        )

        features = features.reshape(
            B,
            S,
            self.feature_dim
        )


        scores = self.slice_attention(
            features
        ).squeeze(-1)


        attention = torch.softmax(
            scores,
            dim=1
        )


        patient_feature = torch.sum(
            features
            *
            attention.unsqueeze(-1),
            dim=1
        )


        return (
            patient_feature,
            attention
        )


    def forward(
        self,
        time1,
        time2,
        clinical,
        return_latent=False
    ):

        F1, _ = self.encode_patient(
            time1
        )

        F2, _ = self.encode_patient(
            time2
        )

        delta_F = (
            F2 - F1
        )


        longitudinal = torch.cat(
            [
                F1,
                F2,
                delta_F
            ],
            dim=1
        )


        mri_feature = self.mri_projection(
            longitudinal
        )

        clinical_feature = self.clinical_projection(
            clinical
        )


        combined = torch.cat(
            [
                mri_feature,
                clinical_feature
            ],
            dim=1
        )


        latent = self.fusion(
            combined
        )

        logits = self.classifier(
            latent
        )


        if return_latent:
            return logits, latent

        return logits


# ============================================================
# LOAD MODEL + PREPROCESSOR
# ============================================================

@st.cache_resource
def load_system():

    model = ImprovedSiameseClinicalCNN(
        clinical_dim=11,
        num_classes=3
    )

    checkpoint = torch.load(
        MODEL_PATH,
        map_location=DEVICE
    )

    model.load_state_dict(
        checkpoint[
            "model_state_dict"
        ]
    )

    model = model.to(
        DEVICE
    )

    model.eval()


    preprocessor = joblib.load(
        PREPROCESSOR_PATH
    )

    return (
        model,
        preprocessor
    )


model, clinical_preprocessor = (
    load_system()
)


# ============================================================
# SIGNED GRAD-CAM
# ============================================================

def signed_gradcam(
    model,
    time1,
    time2,
    clinical
):

    model.eval()

    model.zero_grad(
        set_to_none=True
    )

    captures = []


    def hook_function(
        module,
        inputs,
        output
    ):

        record = {
            "activation":
                output.detach()
        }


        def save_gradient(
            gradient
        ):

            record[
                "gradient"
            ] = gradient.detach()


        output.register_hook(
            save_gradient
        )

        captures.append(
            record
        )


    handle = (
        model.encoder.layer4[-1]
        .register_forward_hook(
            hook_function
        )
    )


    logits = model(
        time1,
        time2,
        clinical
    )

    predicted = int(
        logits.argmax(
            dim=1
        ).item()
    )


    logits[
        0,
        predicted
    ].backward()


    handle.remove()


    def build_cam(
        record
    ):

        activation = record[
            "activation"
        ]

        gradient = record[
            "gradient"
        ]


        weights = gradient.mean(
            dim=(2, 3),
            keepdim=True
        )


        cam = (
            weights
            *
            activation
        ).sum(
            dim=1
        )


        cam = F.interpolate(
            cam.unsqueeze(1),
            size=(224, 224),
            mode="bilinear",
            align_corners=False
        ).squeeze(1)


        return (
            cam.detach()
            .cpu()
            .numpy()
        )


    cam1 = build_cam(
        captures[0]
    )

    cam2 = build_cam(
        captures[1]
    )


    maximum = max(
        np.abs(cam1).max(),
        np.abs(cam2).max(),
        1e-12
    )


    cam1 = np.abs(
        cam1 / maximum
    )

    cam2 = np.abs(
        cam2 / maximum
    )


    return (
        cam1,
        cam2
    )


# ============================================================
# PATIENT METADATA
# ============================================================

metadata = pd.read_csv(
    METADATA_PATH,
    dtype={
        "PatientID": str
    }
)

metadata = metadata.sort_values(
    "PatientID"
).reset_index(
    drop=True
)


# ============================================================
# WEBSITE HEADER
# ============================================================

st.title(
    "🧠 Explainable Longitudinal Glioma Digital Twin"
)

st.caption(
    "Two-timepoint MRI + clinical progression analysis"
)

st.info(
    "Research prototype — not intended for clinical diagnosis."
)


# ============================================================
# PATIENT SELECTION
# ============================================================

st.subheader(
    "Select Patient"
)


patient_id = st.selectbox(
    "Patient ID",
    metadata[
        "PatientID"
    ].tolist()
)


patient_row = metadata[
    metadata["PatientID"]
    ==
    patient_id
].iloc[0]


age = patient_row[
    "Age"
]

sex = patient_row[
    "Sex"
]

diagnosis = patient_row[
    "Diagnosis"
]


# ------------------------------------------------------------
# Auto clinical information
# ------------------------------------------------------------

c1, c2, c3 = st.columns(
    3
)

c1.metric(
    "Age",
    age
)

c2.metric(
    "Sex",
    sex
)

c3.metric(
    "Diagnosis",
    diagnosis
)


# ============================================================
# RUN DIGITAL TWIN
# ============================================================

if st.button(
    "Run Digital Twin Analysis",
    type="primary",
    use_container_width=True
):

    npz_path = os.path.join(
        PROCESSED_DIR,
        f"{patient_id}.npz"
    )


    if not os.path.exists(
        npz_path
    ):

        st.error(
            "Processed MRI file not found in Google Drive."
        )

        st.stop()


    with np.load(
        npz_path
    ) as data:

        time1_np = data[
            "time1"
        ].astype(
            np.float32
        )

        time2_np = data[
            "time2"
        ].astype(
            np.float32
        )

        seg1 = (
            data["seg1"]
            if "seg1" in data.files
            else None
        )

        seg2 = (
            data["seg2"]
            if "seg2" in data.files
            else None
        )


    # ========================================================
    # CLINICAL PREPROCESSING
    # ========================================================

    clinical_df = pd.DataFrame(
        [{
            "Age":
                float(age),

            "Sex":
                sex,

            "Diagnosis":
                diagnosis
        }]
    )


    clinical_np = (
        clinical_preprocessor
        .transform(
            clinical_df
        )
        .astype(
            np.float32
        )
    )


    # ========================================================
    # TORCH INPUT
    # ========================================================

    time1 = torch.from_numpy(
        time1_np
    ).unsqueeze(0).to(
        DEVICE
    )

    time2 = torch.from_numpy(
        time2_np
    ).unsqueeze(0).to(
        DEVICE
    )

    clinical = torch.from_numpy(
        clinical_np
    ).to(
        DEVICE
    )


    # ========================================================
    # PREDICTION
    # ========================================================

    with torch.no_grad():

        logits, latent = model(
            time1,
            time2,
            clinical,
            return_latent=True
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )[0].cpu().numpy()


    predicted_class = int(
        np.argmax(
            probabilities
        )
    )


    # ========================================================
    # FINAL RESULT
    # ========================================================

    st.divider()

    st.header(
        "Digital Twin Result"
    )


    r1, r2, r3 = st.columns(
        3
    )


    r1.metric(
        "Predicted Progression",
        CLASS_NAMES[
            predicted_class
        ]
    )


    r2.metric(
        "Confidence",
        f"{probabilities[predicted_class]*100:.1f}%"
    )


    r3.metric(
        "Research Ground Truth",
        patient_row[
            "Label"
        ]
    )


    # ========================================================
    # CLASS PROBABILITIES
    # ========================================================

    st.subheader(
        "Progression Probabilities"
    )


    probability_df = pd.DataFrame(
        {
            "Class":
                CLASS_NAMES,

            "Probability":
                probabilities
        }
    )


    st.bar_chart(
        probability_df.set_index(
            "Class"
        )
    )


    # ========================================================
    # DIGITAL TWIN LATENT STATE
    # ========================================================

    latent_np = (
        latent[0]
        .cpu()
        .numpy()
    )


    st.subheader(
        "Patient-Specific Digital Twin State"
    )


    st.write(
        "128-dimensional longitudinal latent representation"
    )


    latent_df = pd.DataFrame(
        {
            "Latent Feature":
                np.arange(
                    1,
                    129
                ),

            "Value":
                latent_np
        }
    )


    st.line_chart(
        latent_df.set_index(
            "Latent Feature"
        )
    )


    # ========================================================
    # GRAD-CAM
    # ========================================================

    cam1, cam2 = signed_gradcam(
        model,
        time1,
        time2,
        clinical
    )


    slice_scores = (
        cam1.mean(
            axis=(1, 2)
        )
        +
        cam2.mean(
            axis=(1, 2)
        )
    )


    best_slice = int(
        np.argmax(
            slice_scores
        )
    )


    st.subheader(
        "Longitudinal Explainability"
    )


    g1, g2 = st.columns(
        2
    )


    with g1:

        fig1, ax1 = plt.subplots(
            figsize=(5, 5)
        )

        ax1.imshow(
            time1_np[
                best_slice,
                1
            ],
            cmap="gray"
        )

        ax1.imshow(
            cam1[
                best_slice
            ],
            cmap="jet",
            alpha=0.45
        )

        ax1.set_title(
            "Time1 Signed Grad-CAM"
        )

        ax1.axis(
            "off"
        )

        st.pyplot(
            fig1
        )

        plt.close(
            fig1
        )


    with g2:

        fig2, ax2 = plt.subplots(
            figsize=(5, 5)
        )

        ax2.imshow(
            time2_np[
                best_slice,
                1
            ],
            cmap="gray"
        )

        ax2.imshow(
            cam2[
                best_slice
            ],
            cmap="jet",
            alpha=0.45
        )

        ax2.set_title(
            "Time2 Signed Grad-CAM"
        )

        ax2.axis(
            "off"
        )

        st.pyplot(
            fig2
        )

        plt.close(
            fig2
        )


    # ========================================================
    # TUMOR EVOLUTION
    # ========================================================

    if (
        seg1 is not None
        and
        seg2 is not None
    ):

        et1 = (
            seg1 == 3
        )

        et2 = (
            seg2 == 3
        )


        growth = (
            et2
            &
            (~et1)
        )

        regression = (
            et1
            &
            (~et2)
        )

        stable = (
            et1
            &
            et2
        )


        st.subheader(
            "Tumor Evolution"
        )


        tcol1, tcol2, tcol3 = (
            st.columns(
                3
            )
        )


        tcol1.metric(
            "Growth Voxels",
            f"{int(growth.sum()):,}"
        )

        tcol2.metric(
            "Regression Voxels",
            f"{int(regression.sum()):,}"
        )

        tcol3.metric(
            "Stable ET Voxels",
            f"{int(stable.sum()):,}"
        )


        net_change = (
            int(et2.sum())
            -
            int(et1.sum())
        )


        if net_change > 0:

            evolution_text = (
                f"Net ET increased by "
                f"{net_change:,} voxels."
            )

        elif net_change < 0:

            evolution_text = (
                f"Net ET decreased by "
                f"{abs(net_change):,} voxels."
            )

        else:

            evolution_text = (
                "No net ET volume change."
            )


        st.write(
            evolution_text
        )


    # ========================================================
    # FINAL INTERPRETATION
    # ========================================================

    st.subheader(
        "Final Interpretation"
    )


    st.success(
        f"Digital Twin prediction: "
        f"{CLASS_NAMES[predicted_class]} "
        f"({probabilities[predicted_class]*100:.1f}% confidence)"
    )


    st.warning(
        "Grad-CAM localization showed weak spatial and temporal "
        "alignment with expert ET masks in the held-out cohort. "
        "The maps therefore represent model attention, not precise "
        "tumor segmentation."
    )
'''


app_path = os.path.join(
    APP_DIR,
    "app.py"
)

with open(
    app_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        textwrap.dedent(
            app_code
        )
    )


print(
    "✅ app.py updated for direct Google Drive patient selection."
)


# ============================================================
# 3. INSTALL STREAMLIT + CLOUDFLARED
# ============================================================

subprocess.run(
    "pip install -q streamlit",
    shell=True
)

subprocess.run(
    """
    wget -q \
    https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 \
    -O /content/cloudflared
    """,
    shell=True
)

subprocess.run(
    "chmod +x /content/cloudflared",
    shell=True
)


# ============================================================
# 4. STOP OLD SERVER / TUNNEL
# ============================================================

subprocess.run(
    "pkill -f 'streamlit run app.py' || true",
    shell=True
)

subprocess.run(
    "pkill -f cloudflared || true",
    shell=True
)

time.sleep(2)


# ============================================================
# 5. START WEBSITE
# ============================================================

os.chdir(
    APP_DIR
)

streamlit_process = subprocess.Popen(
    [
        "streamlit",
        "run",
        "app.py",
        "--server.port",
        "8501",
        "--server.headless",
        "true"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(8)


# ============================================================
# 6. CREATE CLICKABLE PUBLIC LINK
# ============================================================

tunnel_process = subprocess.Popen(
    [
        "/content/cloudflared",
        "tunnel",
        "--url",
        "http://127.0.0.1:8501",
        "--no-autoupdate"
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print("\nCreating website link...\n")


website_url = None

for line in tunnel_process.stdout:

    match = re.search(
        r"https://[a-zA-Z0-9-]+\.trycloudflare\.com",
        line
    )

    if match:

        website_url = match.group(0)

        print(
            "======================================"
        )

        print(
            "🧠 DIGITAL TWIN WEBSITE READY"
        )

        print(
            "======================================"
        )

        print()

        print(
            website_url
        )

        print()

        print(
            "✅ Patient ID select করলেই Drive থেকে MRI load হবে."
        )

        break
```

```python
# ============================================================
# FINAL CLASSIFICATION STATISTICS
#
# কী করছি:
#   1. Clinical-only model-এর 95% CI
#   2. Final Improved Siamese+Clinical model-এর 95% CI
#   3. Exact McNemar test
#   4. Report-ready confidence interval graph
#
# IMPORTANT:
#   কোনো model training/tuning হচ্ছে না।
#   শুধুমাত্র held-out test predictions analyse করা হচ্ছে।
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    f1_score
)
from scipy.stats import binomtest


OUTPUT_DIR = os.path.join(
    DATA_DIR,
    "research_outputs"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 1. PREPARE FINAL TEST PREDICTIONS
# ============================================================

y_true = np.asarray(
    y_test
)

clinical_predictions = np.asarray(
    clinical_pred
)

final_predictions = np.asarray(
    final_pred
)

final_true_array = np.asarray(
    final_true
)


# ------------------------------------------------------------
# Safety check:
# McNemar requires predictions from exactly the same patients
# in exactly the same order.
# ------------------------------------------------------------

if not np.array_equal(
    y_true,
    final_true_array
):
    raise ValueError(
        "Test patient order does not match between Clinical-only "
        "and Improved model. Stop here before McNemar analysis."
    )


print(
    "✅ Test patient order verified:",
    len(y_true),
    "patients"
)


# ============================================================
# 2. STRATIFIED PATIENT-LEVEL BOOTSTRAP
#
# কেন stratified:
#   Test set ছোট এবং Decreased class মাত্র 5 patient।
#   তাই bootstrap sample-এ প্রতিটি class preserve করছি।
# ============================================================

def stratified_bootstrap_ci(
    y_true,
    y_pred,
    metric_function,
    n_bootstrap=5000,
    seed=42
):

    rng = np.random.default_rng(
        seed
    )

    classes = np.unique(
        y_true
    )

    bootstrap_scores = []


    for _ in range(
        n_bootstrap
    ):

        sampled_indices = []


        for cls in classes:

            cls_indices = np.where(
                y_true == cls
            )[0]

            sampled_cls = rng.choice(
                cls_indices,
                size=len(cls_indices),
                replace=True
            )

            sampled_indices.extend(
                sampled_cls
            )


        sampled_indices = np.asarray(
            sampled_indices
        )


        score = metric_function(
            y_true[
                sampled_indices
            ],
            y_pred[
                sampled_indices
            ]
        )

        bootstrap_scores.append(
            score
        )


    bootstrap_scores = np.asarray(
        bootstrap_scores
    )


    lower = np.percentile(
        bootstrap_scores,
        2.5
    )

    upper = np.percentile(
        bootstrap_scores,
        97.5
    )


    return (
        lower,
        upper,
        bootstrap_scores
    )


# ============================================================
# 3. METRIC FUNCTIONS
# ============================================================

def accuracy_metric(
    true,
    pred
):
    return accuracy_score(
        true,
        pred
    )


def macro_f1_metric(
    true,
    pred
):
    return f1_score(
        true,
        pred,
        average="macro",
        zero_division=0
    )


# ============================================================
# 4. CLINICAL-ONLY 95% CI
# ============================================================

clinical_accuracy = accuracy_metric(
    y_true,
    clinical_predictions
)

clinical_f1 = macro_f1_metric(
    y_true,
    clinical_predictions
)


clinical_acc_low, clinical_acc_high, _ = (
    stratified_bootstrap_ci(
        y_true,
        clinical_predictions,
        accuracy_metric
    )
)


clinical_f1_low, clinical_f1_high, _ = (
    stratified_bootstrap_ci(
        y_true,
        clinical_predictions,
        macro_f1_metric
    )
)


# ============================================================
# 5. IMPROVED SIAMESE + CLINICAL 95% CI
# ============================================================

final_accuracy = accuracy_metric(
    y_true,
    final_predictions
)

final_f1 = macro_f1_metric(
    y_true,
    final_predictions
)


final_acc_low, final_acc_high, _ = (
    stratified_bootstrap_ci(
        y_true,
        final_predictions,
        accuracy_metric
    )
)


final_f1_low, final_f1_high, _ = (
    stratified_bootstrap_ci(
        y_true,
        final_predictions,
        macro_f1_metric
    )
)


# ============================================================
# 6. EXACT McNEMAR TEST
#
# b = Clinical correct, Final wrong
# c = Clinical wrong, Final correct
#
# H0:
#   Both models have the same error probability.
# ============================================================

clinical_correct = (
    clinical_predictions
    ==
    y_true
)

final_correct = (
    final_predictions
    ==
    y_true
)


b = int(
    np.sum(
        clinical_correct
        &
        (~final_correct)
    )
)

c = int(
    np.sum(
        (~clinical_correct)
        &
        final_correct
    )
)


discordant = (
    b + c
)


if discordant > 0:

    mcnemar_result = binomtest(
        min(b, c),
        n=discordant,
        p=0.5,
        alternative="two-sided"
    )

    mcnemar_p = (
        mcnemar_result.pvalue
    )

else:

    mcnemar_p = 1.0


# ============================================================
# 7. PRINT RESULTS
# ============================================================

print(
    "\n======================================"
)

print(
    "CLASSIFICATION STATISTICAL ANALYSIS"
)

print(
    "======================================"
)


print(
    "\nClinical-only"
)

print(
    f"Accuracy: {clinical_accuracy:.4f} "
    f"[95% CI {clinical_acc_low:.4f}, "
    f"{clinical_acc_high:.4f}]"
)

print(
    f"Macro-F1: {clinical_f1:.4f} "
    f"[95% CI {clinical_f1_low:.4f}, "
    f"{clinical_f1_high:.4f}]"
)


print(
    "\nImproved Siamese + Clinical"
)

print(
    f"Accuracy: {final_accuracy:.4f} "
    f"[95% CI {final_acc_low:.4f}, "
    f"{final_acc_high:.4f}]"
)

print(
    f"Macro-F1: {final_f1:.4f} "
    f"[95% CI {final_f1_low:.4f}, "
    f"{final_f1_high:.4f}]"
)


print(
    "\nExact McNemar Test"
)

print(
    "Clinical correct / Final wrong:",
    b
)

print(
    "Clinical wrong / Final correct:",
    c
)

print(
    f"p-value: {mcnemar_p:.6f}"
)


if mcnemar_p < 0.05:

    print(
        "Result: statistically significant difference."
    )

else:

    print(
        "Result: no statistically significant difference "
        "at alpha = 0.05."
    )


# ============================================================
# 8. SAVE STATISTICAL SUMMARY CSV
# ============================================================

statistics_df = pd.DataFrame(
    [
        {
            "Model":
                "Clinical-only",

            "Accuracy":
                clinical_accuracy,

            "Accuracy CI Low":
                clinical_acc_low,

            "Accuracy CI High":
                clinical_acc_high,

            "Macro-F1":
                clinical_f1,

            "Macro-F1 CI Low":
                clinical_f1_low,

            "Macro-F1 CI High":
                clinical_f1_high
        },

        {
            "Model":
                "Improved Siamese + Clinical",

            "Accuracy":
                final_accuracy,

            "Accuracy CI Low":
                final_acc_low,

            "Accuracy CI High":
                final_acc_high,

            "Macro-F1":
                final_f1,

            "Macro-F1 CI Low":
                final_f1_low,

            "Macro-F1 CI High":
                final_f1_high
        }
    ]
)


stats_csv = os.path.join(
    OUTPUT_DIR,
    "classification_statistical_summary.csv"
)


statistics_df.to_csv(
    stats_csv,
    index=False
)


# ============================================================
# 9. SAVE McNEMAR RESULT
# ============================================================

mcnemar_df = pd.DataFrame(
    [
        {
            "Clinical correct / Final wrong":
                b,

            "Clinical wrong / Final correct":
                c,

            "Discordant pairs":
                discordant,

            "Exact McNemar p-value":
                mcnemar_p
        }
    ]
)


mcnemar_csv = os.path.join(
    OUTPUT_DIR,
    "clinical_vs_final_mcnemar.csv"
)


mcnemar_df.to_csv(
    mcnemar_csv,
    index=False
)


# ============================================================
# 10. REPORT-READY 95% CI FIGURE
# ============================================================

models = [
    "Clinical-only",
    "Improved Siamese + Clinical"
]


accuracy_scores = np.array(
    [
        clinical_accuracy,
        final_accuracy
    ]
)


accuracy_lower = np.array(
    [
        clinical_acc_low,
        final_acc_low
    ]
)


accuracy_upper = np.array(
    [
        clinical_acc_high,
        final_acc_high
    ]
)


f1_scores = np.array(
    [
        clinical_f1,
        final_f1
    ]
)


f1_lower = np.array(
    [
        clinical_f1_low,
        final_f1_low
    ]
)


f1_upper = np.array(
    [
        clinical_f1_high,
        final_f1_high
    ]
)


x = np.arange(
    len(models)
)

offset = 0.10


plt.figure(
    figsize=(9, 6)
)


plt.errorbar(
    x - offset,
    accuracy_scores,
    yerr=[
        accuracy_scores
        -
        accuracy_lower,

        accuracy_upper
        -
        accuracy_scores
    ],
    fmt="o",
    capsize=6,
    label="Accuracy"
)


plt.errorbar(
    x + offset,
    f1_scores,
    yerr=[
        f1_scores
        -
        f1_lower,

        f1_upper
        -
        f1_scores
    ],
    fmt="o",
    capsize=6,
    label="Macro-F1"
)


plt.xticks(
    x,
    models
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1
)

plt.title(
    "Held-Out Test Performance with 95% Bootstrap Confidence Intervals"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.legend()

plt.tight_layout()


figure_path = os.path.join(
    OUTPUT_DIR,
    "classification_95ci_comparison.png"
)


plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. FINAL SAVE MESSAGE
# ============================================================

print(
    "\n✅ Statistical summary saved:"
)

print(
    stats_csv
)

print(
    "\n✅ McNemar result saved:"
)

print(
    mcnemar_csv
)

print(
    "\n✅ Report-ready CI graph saved:"
)

print(
    figure_path
)

print(
    "\n🔒 Classification statistical analysis complete."
)
```
